# BharatCrop Intelligence
## Notebook 01: Project Definition and Data-Source Registry

### Objectives
- Define the objectives and scope of the project.
- Identify the official datasets required for nationwide agricultural analysis.
- Record source ownership, access methods, geographic coverage, temporal coverage, and licensing.
- Establish a consistent process for verifying datasets before using them.
- Prepare the foundation for data acquisition, preprocessing, analysis, and machine learning.

### Working principle
No dataset will be considered verified until its actual contents,
coverage, schema, and access conditions have been examined.

```python

# Check that each registered source has a unique identifier.
assert source_registry["source_id"].is_unique

# Check that essential source information is present.
assert source_registry["dataset_name"].notna().all()
assert source_registry["organization"].notna().all()
assert source_registry["official_url"].notna().all()

# Confirm that no source has been marked verified prematurely.
verification_columns = [
    "schema_verified",
    "download_verified",
    "coverage_verified",
    "license_verified",
    "integration_verified",
]

assert not source_registry[verification_columns].any().any()

# Save the registry and data requirements for future notebooks.
metadata_dir = PROJECT_ROOT / "metadata"
metadata_dir.mkdir(parents=True, exist_ok=True)

registry_path = metadata_dir / "source_registry.csv"
requirements_path = metadata_dir / "data_requirements.csv"

source_registry.to_csv(registry_path, index=False)
data_requirements.to_csv(requirements_path, index=False)

print("Registry validation passed.")
print(f"Registered sources: {len(source_registry)}")
print(f"Verified sources: {int(source_registry['download_verified'].sum())}")
print(f"Registry saved: {registry_path.resolve()}")
print(f"Requirements saved: {requirements_path.resolve()}")
```
