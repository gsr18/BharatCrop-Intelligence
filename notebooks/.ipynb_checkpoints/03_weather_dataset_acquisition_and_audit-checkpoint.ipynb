## Candidate Weather Source: NASA POWER

Objective:
Test whether NASA POWER can provide usable daily weather data in CSV format.

Pilot scope:
- One test location
- Calendar year 2013
- Daily precipitation, maximum temperature, and minimum temperature
- Preserve the downloaded CSV as raw data
- Inspect date coverage, units, missing values, and file integrity

Important:
This is a source evaluation pilot, not the final nationwide dataset.
NASA POWER estimates are not identical to IMD's gridded observations.


```python

# Cell 6: Download a one-year NASA POWER weather pilot as CSV.
# This does not download nationwide data or modify the APY dataset.

from pathlib import Path
import requests

# Reuse PROJECT_ROOT and RAW_DIR from earlier notebook cells.
pilot_dir = RAW_DIR / "weather_pilot"
pilot_dir.mkdir(parents=True, exist_ok=True)

# Test point near Ludhiana, Punjab.
latitude = 30.90
longitude = 75.85

start_date = "20130101"
end_date = "20131231"

parameters = "PRECTOTCORR,T2M_MAX,T2M_MIN"

url = "https://power.larc.nasa.gov/api/temporal/daily/point"

params = {
    "parameters": parameters,
    "community": "AG",
    "longitude": longitude,
    "latitude": latitude,
    "start": start_date,
    "end": end_date,
    "format": "CSV",
    "time-standard": "UTC",
}

response = requests.get(url, params=params, timeout=120)

print("HTTP status:", response.status_code)
response.raise_for_status()

# Save the provider's original CSV response without editing it.
csv_path = pilot_dir / "nasa_power_weather_pilot_2013.csv"

csv_path.write_bytes(response.content)

print("Saved:", csv_path)
print("File size (bytes):", csv_path.stat().st_size)
print("Content type:", response.headers.get("Content-Type"))
print("\nFirst 20 lines of response:")
print("\n".join(response.text.splitlines()[:20]))
```

```python

# Cell 7: Inspect the saved response and locate its data table.
# We inspect the format before assuming where the CSV header begins.

lines = csv_path.read_text(encoding="utf-8-sig").splitlines()

for i, line in enumerate(lines[:40]):
    print(f"{i + 1:02d}: {line}")

print("\nTotal response lines:", len(lines))
```

```python

# Cell 8: Parse the NASA POWER pilot CSV and perform a basic audit.
# The original downloaded file remains unchanged.

import pandas as pd
import numpy as np

# NASA POWER places metadata before the actual CSV table.
header_index = next(
    i for i, line in enumerate(lines)
    if line.startswith("YEAR,DOY,")
)

weather = pd.read_csv(csv_path, skiprows=header_index)

# Convert year and day-of-year into proper calendar dates.
weather["DATE"] = pd.to_datetime(
    weather["YEAR"].astype(str)
    + weather["DOY"].astype(str).str.zfill(3),
    format="%Y%j",
    errors="raise",
)

weather = weather[
    ["DATE", "PRECTOTCORR", "T2M_MAX", "T2M_MIN"]
].copy()

# NASA POWER's documented missing-value sentinel is -999.
parameters = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]
weather[parameters] = weather[parameters].replace(-999, np.nan)

print("Dataset shape:", weather.shape)
print("Date range:", weather["DATE"].min(), "to", weather["DATE"].max())
print("Duplicate dates:", weather["DATE"].duplicated().sum())
print("\nMissing values by variable:")
print(weather[parameters].isna().sum())
print("\nDescriptive statistics:")
display(weather[parameters].describe().round(2))

print("\nFirst 5 parsed rows:")
display(weather.head())
```

```python
# Cell 9: Structural and basic physical-consistency checks.
# These checks flag issues; they do not automatically remove records.

expected_dates = pd.date_range("2013-01-01", "2013-12-31", freq="D")

print("Expected days:", len(expected_dates))
print("Actual rows:", len(weather))
print("Unique dates:", weather["DATE"].nunique())
print("Missing calendar dates:",
      len(set(expected_dates) - set(weather["DATE"])))
print("Negative precipitation values:",
      (weather["PRECTOTCORR"] < 0).sum())
print("Days where T2M_MIN > T2M_MAX:",
      (weather["T2M_MIN"] > weather["T2M_MAX"]).sum())

print("\nSource metadata:")
print("Precipitation: corrected precipitation, mm/day")
print("Temperature: daily maximum and minimum, degrees Celsius")
print("Source: NASA POWER, MERRA-2")
```

```python

# Cell 10: Save the preliminary NASA POWER pilot audit.
# This report describes only the 2013 test point.

import json
from datetime import datetime, timezone

audit_summary = {
    "project": "BharatCrop Intelligence",
    "source": "NASA POWER",
    "dataset": "Daily point weather pilot",
    "year": 2013,
    "latitude": latitude,
    "longitude": longitude,
    "variables": {
        "PRECTOTCORR": "mm/day",
        "T2M_MAX": "degrees Celsius",
        "T2M_MIN": "degrees Celsius",
    },
    "rows": int(len(weather)),
    "unique_dates": int(weather["DATE"].nunique()),
    "start_date": weather["DATE"].min().strftime("%Y-%m-%d"),
    "end_date": weather["DATE"].max().strftime("%Y-%m-%d"),
    "missing_values": {
        column: int(weather[column].isna().sum())
        for column in parameters
    },
    "negative_precipitation_count": int(
        (weather["PRECTOTCORR"] < 0).sum()
    ),
    "minimum_greater_than_maximum_count": int(
        (weather["T2M_MIN"] > weather["T2M_MAX"]).sum()
    ),
    "source_verified_end_to_end": False,
    "nationwide_coverage_verified": False,
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
}

audit_path = AUDIT_DIR / "nasa_power_weather_pilot_audit.json"

with open(audit_path, "w", encoding="utf-8") as file:
    json.dump(audit_summary, file, indent=2)

print("Pilot audit saved:", audit_path)
print("Status: PRELIMINARY PILOT PASSED")
print("Nationwide acquisition: NOT STARTED")
```

```python

import pandas as pd

# Work on the attribute table; retain the original geometry dataset unchanged.
boundary_attributes = boundaries.drop(columns="geometry").copy()

# Check the completeness of key administrative fields.
for column in ["STATE_UT", "STATE_LGD", "DISTRICT", "DIST_LGD"]:
    boundary_attributes[column] = (
        boundary_attributes[column].astype("string").str.strip()
    )

print("=== Boundary identifier audit ===")
print("Total features:", len(boundary_attributes))
print("Unique state labels:", boundary_attributes["STATE_UT"].nunique())
print("Unique district labels:", boundary_attributes["DISTRICT"].nunique())

print("\nMissing or unavailable identifiers:")
for column in ["STATE_UT", "STATE_LGD", "DISTRICT", "DIST_LGD"]:
    values = boundary_attributes[column]
    unavailable = values.isna() | values.str.upper().isin(
        ["", "NOT AVAILABLE", "NA", "N/A"]
    )
    print(f"{column}: {int(unavailable.sum())}")

# Check repeated state–district name pairs.
duplicate_pairs = boundary_attributes[
    boundary_attributes.duplicated(
        subset=["STATE_UT", "DISTRICT"], keep=False
    )
].sort_values(["STATE_UT", "DISTRICT"])

print("\nRepeated state–district name pairs:", len(duplicate_pairs))
if not duplicate_pairs.empty:
    display(
        duplicate_pairs[
            ["STATE_UT", "STATE_LGD", "DISTRICT", "DIST_LGD", "REMARKS"]
        ].head(30)
    )

# Summarize features by state/UT as recorded in the source.
print("\nFeatures per state/UT:")
display(
    boundary_attributes.groupby("STATE_UT", dropna=False)
    .size()
    .sort_values(ascending=False)
    .rename("boundary_feature_count")
    .to_frame()
)

# Inspect records with unavailable district LGD codes.
unavailable_district_codes = boundary_attributes[
    boundary_attributes["DIST_LGD"].isna()
    | boundary_attributes["DIST_LGD"].str.upper().isin(
        ["", "NOT AVAILABLE", "NA", "N/A"]
    )
]

print("\nRecords with unavailable district LGD codes:")
display(
    unavailable_district_codes[
        ["STATE_UT", "STATE_LGD", "DISTRICT", "DIST_LGD", "REMARKS"]
    ].head(30)
)
```

```python

import json
from datetime import datetime, timezone

# Identify records with missing or explicitly unavailable identifiers.
key_columns = ["STATE_UT", "STATE_LGD", "DISTRICT", "DIST_LGD"]

unavailable_counts = {}
for column in key_columns:
    values = boundary_attributes[column].astype("string").str.strip()
    unavailable = values.isna() | values.str.upper().isin(
        ["", "NOT AVAILABLE", "NA", "N/A"]
    )
    unavailable_counts[column] = int(unavailable.sum())

# Preserve records with missing administrative identifiers for review.
unresolved = boundary_attributes[
    boundary_attributes[key_columns].isna().any(axis=1)
    | boundary_attributes[key_columns].apply(
        lambda col: col.astype("string").str.upper().isin(
            ["", "NOT AVAILABLE", "NA", "N/A"]
        )
    ).any(axis=1)
].copy()

unresolved_path = AUDIT_DIR / "district_boundary_unresolved_records.csv"
unresolved.to_csv(unresolved_path, index=False)

boundary_audit = {
    "source_file": "DISTRICT_BOUNDARY.shp",
    "feature_count": int(len(boundaries)),
    "state_label_count": int(boundary_attributes["STATE_UT"].nunique()),
    "unique_district_label_count": int(boundary_attributes["DISTRICT"].nunique()),
    "missing_or_unavailable_identifiers": unavailable_counts,
    "missing_geometry_count": int(boundaries.geometry.isna().sum()),
    "empty_geometry_count": int(boundaries.geometry.is_empty.sum()),
    "crs": str(boundaries.crs),
    "unresolved_record_count": int(len(unresolved)),
    "matching_status": "NOT_STARTED",
    "created_at_utc": datetime.now(timezone.utc).isoformat(),
}

audit_path = AUDIT_DIR / "district_boundary_audit.json"
with open(audit_path, "w", encoding="utf-8") as file:
    json.dump(boundary_audit, file, indent=2)

print("Boundary audit saved:", audit_path)
print("Unresolved records saved:", unresolved_path)
print("Unresolved records retained:", len(unresolved))
print("APY-to-boundary matching: NOT STARTED")
```

```python

import re
import pandas as pd

# Load the APY geography reference created earlier.
APY_REFERENCE_PATH = (
    PROJECT_ROOT / "metadata" / "apy_district_geographic_reference.csv"
)

apy_geo = pd.read_csv(
    APY_REFERENCE_PATH,
    dtype={"State": "string", "District": "string"},
)

# Copy boundary attributes so the original GeoDataFrame remains unchanged.
boundary_geo = boundary_attributes.copy()

def normalize_name(value):
    """Normalize spacing and case without guessing geographic equivalence."""
    if pd.isna(value):
        return None
    value = str(value).strip().upper()
    value = re.sub(r"\s+", " ", value)
    return value if value else None

# Create comparison keys; preserve original names in the source columns.
apy_geo["state_key"] = apy_geo["State"].map(normalize_name)
apy_geo["district_key"] = apy_geo["District"].map(normalize_name)

boundary_geo["state_key"] = boundary_geo["STATE_UT"].map(normalize_name)
boundary_geo["district_key"] = boundary_geo["DISTRICT"].map(normalize_name)

# Only uniquely occurring state–district pairs are eligible for this
# preliminary exact-name comparison.
boundary_pair_counts = boundary_geo.groupby(
    ["state_key", "district_key"], dropna=False
).size()

unique_boundary_pairs = boundary_pair_counts[boundary_pair_counts == 1].index

boundary_geo["unique_boundary_pair"] = pd.MultiIndex.from_frame(
    boundary_geo[["state_key", "district_key"]]
).isin(unique_boundary_pairs)

# Build a lookup only from unique, fully identified pairs.
valid_boundary = boundary_geo[
    boundary_geo["state_key"].notna()
    & boundary_geo["district_key"].notna()
    & boundary_geo["unique_boundary_pair"]
].copy()

valid_keys = set(zip(valid_boundary["state_key"], valid_boundary["district_key"]))

apy_geo["exact_name_candidate"] = [
    (state, district) in valid_keys
    for state, district in zip(apy_geo["state_key"], apy_geo["district_key"])
]

apy_geo["boundary_match_status"] = apy_geo["exact_name_candidate"].map(
    {True: "EXACT_NAME_CANDIDATE_REQUIRES_REVIEW",
     False: "UNMATCHED"}
)

print("APY geography rows:", len(apy_geo))
print("Exact-name candidates:", int(apy_geo["exact_name_candidate"].sum()))
print("Unmatched APY rows:", int((~apy_geo["exact_name_candidate"]).sum()))

print("\nCandidate matches for review:")
display(
    apy_geo.loc[
        apy_geo["exact_name_candidate"],
        ["apy_geography_id", "State", "District", "boundary_match_status"]
    ].head(25)
)

print("\nSample unmatched APY geography:")
display(
    apy_geo.loc[
        ~apy_geo["exact_name_candidate"],
        ["apy_geography_id", "State", "District"]
    ].head(25)
)
```

```python

# ============================================================
# CELL 19: GENERATE UNMATCHED DISTRICT CANDIDATES
# ============================================================

from pathlib import Path
from difflib import SequenceMatcher
import pandas as pd

# ------------------------------------------------------------
# STEP 1: Prepare the boundary district reference
# ------------------------------------------------------------

boundary_pairs = boundary_attributes[
    ["STATE_UT", "DISTRICT"]
].dropna().copy()

# Normalize names for comparison without changing original names.
boundary_pairs["state_key"] = (
    boundary_pairs["STATE_UT"]
    .astype(str)
    .str.strip()
    .str.upper()
)

boundary_pairs["district_key"] = (
    boundary_pairs["DISTRICT"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Remove blank and unavailable district names.
invalid_names = {"", "NA", "N/A", "NOT AVAILABLE"}

boundary_pairs = boundary_pairs[
    (~boundary_pairs["state_key"].isin(invalid_names))
    & (~boundary_pairs["district_key"].isin(invalid_names))
].copy()

# ------------------------------------------------------------
# STEP 2: Exclude ambiguous repeated state-district pairs
# ------------------------------------------------------------

pair_counts = (
    boundary_pairs
    .groupby(["state_key", "district_key"])
    .size()
)

unique_pair_index = pair_counts[
    pair_counts == 1
].index

boundary_pairs = (
    boundary_pairs
    .set_index(["state_key", "district_key"])
    .loc[lambda df: df.index.isin(unique_pair_index)]
    .reset_index()
)

# ------------------------------------------------------------
# STEP 3: Identify unmatched APY districts
# ------------------------------------------------------------

unmatched = apy_geo.loc[
    ~apy_geo["exact_name_candidate"],
    ["apy_geography_id", "State", "District"]
].copy()

unmatched["state_key"] = (
    unmatched["State"]
    .astype(str)
    .str.strip()
    .str.upper()
)

unmatched["district_key"] = (
    unmatched["District"]
    .astype(str)
    .str.strip()
    .str.upper()
)

print("Unmatched APY districts:", len(unmatched))
print("Usable boundary district pairs:", len(boundary_pairs))

# ------------------------------------------------------------
# STEP 4: Find the three most similar names within each state
# ------------------------------------------------------------

suggestions = []

for _, row in unmatched.iterrows():

    # Compare districts only within the same state.
    same_state = boundary_pairs[
        boundary_pairs["state_key"] == row["state_key"]
    ]

    scored_candidates = []

    for _, boundary_row in same_state.iterrows():

        score = SequenceMatcher(
            None,
            row["district_key"],
            boundary_row["district_key"]
        ).ratio()

        scored_candidates.append({
            "candidate_boundary_district": boundary_row["DISTRICT"],
            "similarity_score": round(score, 3)
        })

    # Rank candidates from highest to lowest similarity.
    scored_candidates.sort(
        key=lambda item: item["similarity_score"],
        reverse=True
    )

    top = scored_candidates[:3]

    # Keep records even when no candidate exists in that state.
    while len(top) < 3:
        top.append({
            "candidate_boundary_district": None,
            "similarity_score": None
        })

    suggestions.append({
        "apy_geography_id": row["apy_geography_id"],
        "State": row["State"],
        "APY_District": row["District"],

        "candidate_1": top[0]["candidate_boundary_district"],
        "score_1": top[0]["similarity_score"],

        "candidate_2": top[1]["candidate_boundary_district"],
        "score_2": top[1]["similarity_score"],

        "candidate_3": top[2]["candidate_boundary_district"],
        "score_3": top[2]["similarity_score"],

        # No candidate is automatically approved.
        "review_status": "NOT_REVIEWED"
    })

# Create the review DataFrame.
match_review = pd.DataFrame(suggestions)

# ------------------------------------------------------------
# STEP 5: Create the output folder and save the CSV
# ------------------------------------------------------------

audit_dir = Path("reports") / "data_audit"
audit_dir.mkdir(parents=True, exist_ok=True)

review_path = (
    audit_dir / "apy_unmatched_district_candidate_review.csv"
)

match_review.to_csv(review_path, index=False)

# ------------------------------------------------------------
# STEP 6: Verify and display the results
# ------------------------------------------------------------

print("\nCandidate generation completed.")
print("Total unmatched records:", len(match_review))
print("Output file:", review_path.resolve())
print("File exists:", review_path.exists())

print("\nFirst 20 candidate suggestions:")
display(
    match_review
    .sort_values(["State", "APY_District"])
    .head(20)
    .reset_index(drop=True)
)

print("\nUnmatched district counts by state:")
display(
    match_review
    .groupby("State")
    .size()
    .sort_values(ascending=False)
    .rename("unmatched_districts")
    .to_frame()
)
```

```python

# ============================================================
# CELL 21: CORRECTED DISTRICT MATCH REVIEW
# ============================================================

from pathlib import Path
from difflib import SequenceMatcher
import pandas as pd
import re


def normalize_district_name(value):
    """Create a comparison key without changing the source name."""
    if pd.isna(value):
        return ""

    name = str(value).upper().strip()

    # Working hypothesis for the apparent character corruption.
    name = name.replace(">", "A")

    # Normalize punctuation and spaces.
    name = name.replace("&", " AND ")
    name = re.sub(r"[^A-Z0-9]+", " ", name)
    name = re.sub(r"\s+", " ", name).strip()

    return name


# 1. Prepare boundary pairs and preserve original names.
boundary_pairs = boundary_attributes[
    ["STATE_UT", "DISTRICT", "DIST_LGD"]
].dropna(subset=["STATE_UT", "DISTRICT"]).copy()

boundary_pairs["state_key"] = (
    boundary_pairs["STATE_UT"].map(normalize_district_name)
)
boundary_pairs["district_key"] = (
    boundary_pairs["DISTRICT"].map(normalize_district_name)
)

invalid_names = {"", "NA", "N A", "NOT AVAILABLE"}

boundary_pairs = boundary_pairs[
    (~boundary_pairs["state_key"].isin(invalid_names))
    & (~boundary_pairs["district_key"].isin(invalid_names))
].copy()

# Exclude ambiguous repeated state-district pairs.
pair_counts = boundary_pairs.groupby(
    ["state_key", "district_key"]
).size()

unique_pair_index = pair_counts[
    pair_counts == 1
].index

boundary_pairs = (
    boundary_pairs
    .set_index(["state_key", "district_key"])
    .loc[lambda df: df.index.isin(unique_pair_index)]
    .reset_index()
)

# 2. Prepare the APY records previously marked unmatched.
unmatched = apy_geo.loc[
    ~apy_geo["exact_name_candidate"],
    ["apy_geography_id", "State", "District"]
].copy()

unmatched["state_key"] = (
    unmatched["State"].map(normalize_district_name)
)
unmatched["district_key"] = (
    unmatched["District"].map(normalize_district_name)
)

# 3. Generate candidate matches using normalized names on both sides.
review_rows = []

for _, row in unmatched.iterrows():

    same_state = boundary_pairs[
        boundary_pairs["state_key"] == row["state_key"]
    ].copy()

    same_state["similarity_score"] = same_state[
        "district_key"
    ].apply(
        lambda candidate_name: SequenceMatcher(
            None,
            row["district_key"],
            candidate_name
        ).ratio()
    )

    same_state = same_state.sort_values(
        "similarity_score",
        ascending=False
    )

    exact_candidates = same_state[
        same_state["district_key"] == row["district_key"]
    ]

    top = same_state.head(3).to_dict("records")

    while len(top) < 3:
        top.append({
            "DISTRICT": None,
            "DIST_LGD": None,
            "district_key": None,
            "similarity_score": None
        })

    if len(exact_candidates) == 1:
        status = "EXACT_NORMALIZED_CANDIDATE_REQUIRES_REVIEW"
    elif len(exact_candidates) > 1:
        status = "AMBIGUOUS_EXACT_CANDIDATE"
    else:
        status = "FUZZY_CANDIDATE_REQUIRES_REVIEW"

    review_rows.append({
        "apy_geography_id": row["apy_geography_id"],
        "State": row["State"],
        "APY_District": row["District"],
        "APY_normalized_name": row["district_key"],
        "candidate_1": top[0]["DISTRICT"],
        "candidate_1_normalized": top[0]["district_key"],
        "candidate_1_lgd": top[0]["DIST_LGD"],
        "score_1": (
            round(top[0]["similarity_score"], 3)
            if top[0]["similarity_score"] is not None
            else None
        ),
        "candidate_2": top[1]["DISTRICT"],
        "candidate_2_lgd": top[1]["DIST_LGD"],
        "score_2": (
            round(top[1]["similarity_score"], 3)
            if top[1]["similarity_score"] is not None
            else None
        ),
        "candidate_3": top[2]["DISTRICT"],
        "candidate_3_lgd": top[2]["DIST_LGD"],
        "score_3": (
            round(top[2]["similarity_score"], 3)
            if top[2]["similarity_score"] is not None
            else None
        ),
        "review_status": status
    })

match_review_corrected = pd.DataFrame(review_rows)

# 4. Save to the same audit directory.
audit_dir = Path("reports") / "data_audit"
audit_dir.mkdir(parents=True, exist_ok=True)

output_path = (
    audit_dir / "apy_unmatched_district_candidate_review_corrected.csv"
)

match_review_corrected.to_csv(output_path, index=False)

# 5. Summarize results.
print("Records reviewed:", len(match_review_corrected))
print(
    match_review_corrected["review_status"]
    .value_counts(dropna=False)
)

print("\nExact normalized candidates requiring review:")
display(
    match_review_corrected[
        match_review_corrected["review_status"]
        == "EXACT_NORMALIZED_CANDIDATE_REQUIRES_REVIEW"
    ][
        [
            "apy_geography_id",
            "State",
            "APY_District",
            "candidate_1",
            "candidate_1_lgd",
            "score_1",
            "review_status"
        ]
    ].reset_index(drop=True)
)

print("\nSaved to:", output_path.resolve())
```

```python
# ============================================================
# BharatCrop Intelligence — District Coordinate Crosswalk
# Purpose:
#   1. Generate representative points for boundary matches.
#   2. Use explicit alias mappings for known name differences.
#   3. Fill unmatched districts with deterministic synthetic
#      coordinates, clearly marked as synthetic.
#   4. Save a traceable crosswalk without modifying source data.
# ============================================================

import pandas as pd
import numpy as np
import geopandas as gpd
import re
import json
from pathlib import Path

# ------------------------------------------------------------
# 1. Set project paths
# ------------------------------------------------------------
project_root = Path.cwd().parent
metadata_dir = project_root / "metadata"
reports_dir = project_root / "reports" / "data_audit"

metadata_dir.mkdir(parents=True, exist_ok=True)
reports_dir.mkdir(parents=True, exist_ok=True)

output_path = metadata_dir / "apy_district_coordinate_crosswalk.csv"
summary_path = reports_dir / "apy_coordinate_crosswalk_summary.json"

# ------------------------------------------------------------
# 2. Normalize names for matching
# ------------------------------------------------------------
def normalize_geo_name(value):
    if pd.isna(value):
        return ""
    value = str(value).upper().strip()
    value = value.replace(">", "A")  # Known suspected source-text corruption
    value = value.replace("&", " AND ")
    value = re.sub(r"[^A-Z0-9]+", " ", value)
    return re.sub(r"\s+", " ", value).strip()


apy = apy_geo.copy().reset_index(drop=True)

apy["state_key"] = apy["State"].map(normalize_geo_name)
apy["district_key"] = apy["District"].map(normalize_geo_name)

# Convert boundaries to WGS84 so coordinates are latitude/longitude.
boundaries_wgs84 = boundaries.to_crs("EPSG:4326").copy()

boundaries_wgs84["state_key"] = (
    boundaries_wgs84["STATE_UT"].map(normalize_geo_name)
)
boundaries_wgs84["district_key"] = (
    boundaries_wgs84["DISTRICT"].map(normalize_geo_name)
)

# ------------------------------------------------------------
# 3. Explicit aliases for common spelling/name variations
#    These are candidates, not proof of historical equivalence.
# ------------------------------------------------------------
district_aliases = {
    ("MEGHALAYA", "RI BHOI"): "RI BHOI",
    ("JHARKHAND", "SARAIKELA KHARSAWAN"): "SARAIKELA KHARSAWAN",
    ("MADHYA PRADESH", "AGAR MALWA"): "AGAR MALWA",
    ("CHHATTISGARH", "JANJGIR CHAMPA"): "JANJGIR CHAMPA",
    ("TELANGANA", "MEDCHAL MALKAJGIRI"): "MEDCHAL MALKAJGIRI",
}

state_aliases = {
    "KERALA": "KERALAM",
    "DADRA AND NAGAR HAVELI AND DAMAN AND DIU":
        "DADRA AND NAGAR HAVELI AND DAMAN AND DIU",
}

# Unique state/district polygons only. Ambiguous duplicates are excluded.
boundary_counts = (
    boundaries_wgs84.groupby(["state_key", "district_key"])
    .size()
    .rename("boundary_count")
    .reset_index()
)

unique_keys = boundary_counts.loc[
    boundary_counts["boundary_count"] == 1,
    ["state_key", "district_key"]
]

unique_boundaries = boundaries_wgs84.merge(
    unique_keys,
    on=["state_key", "district_key"],
    how="inner"
)

# Build a lookup using representative points inside each polygon.
boundary_lookup = {}

for _, row in unique_boundaries.iterrows():
    point = row.geometry.representative_point()
    boundary_lookup[(row["state_key"], row["district_key"])] = {
        "latitude": float(point.y),
        "longitude": float(point.x),
        "matched_boundary_name": str(row["DISTRICT"]),
        "boundary_state_name": str(row["STATE_UT"]),
        "district_lgd": str(row["DIST_LGD"]),
        "state_lgd": str(row["STATE_LGD"]),
    }

# ------------------------------------------------------------
# 4. Match APY districts to boundary representatives
# ------------------------------------------------------------
results = []

for _, row in apy.iterrows():
    state = row["State"]
    district = row["District"]
    state_key = row["state_key"]
    district_key = row["district_key"]

    # State alias support: try original state key first.
    candidate_state_keys = [state_key]

    if state_key == "KERALA":
        candidate_state_keys.append("KERALAM")

    match_key = (state_key, district_key)

    # First try direct normalized matching.
    if match_key not in boundary_lookup:
        # Try a state/district alias candidate.
        alias_district_key = district_aliases.get(
            (state_key, district_key)
        )
        if alias_district_key:
            for candidate_state in candidate_state_keys:
                candidate_key = (candidate_state, alias_district_key)
                if candidate_key in boundary_lookup:
                    match_key = candidate_key
                    break

    match = boundary_lookup.get(match_key)

    if match:
        results.append({
            "apy_geography_id": row["apy_geography_id"],
            "State": state,
            "District": district,
            "latitude": match["latitude"],
            "longitude": match["longitude"],
            "coordinate_source": "BOUNDARY_REPRESENTATIVE_POINT",
            "is_synthetic": False,
            "geography_confidence": "PROVISIONAL_BOUNDARY_MATCH",
            "matched_boundary_name": match["matched_boundary_name"],
            "state_lgd": match["state_lgd"],
            "district_lgd": match["district_lgd"],
            "match_notes": (
                "Representative point from boundary polygon. "
                "Historical district equivalence not yet verified."
            ),
        })
    else:
        results.append({
            "apy_geography_id": row["apy_geography_id"],
            "State": state,
            "District": district,
            "latitude": np.nan,
            "longitude": np.nan,
            "coordinate_source": "PENDING_SYNTHETIC",
            "is_synthetic": True,
            "geography_confidence": "LOW_SYNTHETIC_PLACEHOLDER",
            "matched_boundary_name": "",
            "state_lgd": "",
            "district_lgd": "",
            "match_notes": (
                "No unique normalized boundary match. "
                "Synthetic coordinate assignment is pending."
            ),
        })

crosswalk = pd.DataFrame(results)

# ------------------------------------------------------------
# 5. Generate deterministic placeholder coordinates
# ------------------------------------------------------------
# Important: these are NOT real district locations and should
# not be used for geographic analysis or production weather
# extraction. They are temporary, labelled placeholders only.
#
# We assign them near the approximate state-level anchor.
# The offsets are deterministic, so repeated runs are stable.

state_anchor = {
    "ANDHRA PRADESH": (16.5, 80.6),
    "ARUNACHAL PRADESH": (28.0, 94.5),
    "ASSAM": (26.2, 92.9),
    "BIHAR": (25.8, 85.1),
    "CHHATTISGARH": (21.3, 82.0),
    "GOA": (15.4, 74.0),
    "GUJARAT": (22.3, 71.8),
    "HARYANA": (29.0, 76.0),
    "HIMACHAL PRADESH": (31.8, 77.2),
    "JHARKHAND": (23.6, 85.5),
    "KARNATAKA": (15.0, 76.0),
    "KERALA": (10.5, 76.3),
    "KERALAM": (10.5, 76.3),
    "MADHYA PRADESH": (23.5, 78.5),
    "MAHARASHTRA": (19.5, 75.5),
    "MANIPUR": (24.7, 93.9),
    "MEGHALAYA": (25.5, 91.3),
    "MIZORAM": (23.2, 92.7),
    "NAGALAND": (26.0, 94.5),
    "ODISHA": (20.5, 84.5),
    "PUNJAB": (31.0, 75.5),
    "RAJASTHAN": (26.5, 73.8),
    "SIKKIM": (27.5, 88.5),
    "TAMIL NADU": (11.0, 78.5),
    "TELANGANA": (17.8, 79.0),
    "TRIPURA": (23.8, 91.5),
    "UTTAR PRADESH": (27.0, 80.5),
    "UTTARAKHAND": (30.0, 79.0),
    "WEST BENGAL": (23.5, 87.5),
    "JAMMU AND KASHMIR": (33.5, 75.0),
    "LADAKH": (34.2, 77.6),
    "PUDUCHERRY": (11.9, 79.8),
    "CHANDIGARH": (30.7, 76.8),
    "ANDAMAN AND NICOBAR ISLANDS": (11.7, 92.7),
    "DADRA AND NAGAR HAVELI AND DAMAN AND DIU": (20.2, 73.0),
}

# Stable small offsets, used only to make placeholder points distinct.
for state_key, group in crosswalk[
    crosswalk["is_synthetic"]
].groupby(crosswalk["State"].map(normalize_geo_name)):
    anchor = state_anchor.get(state_key)

    if anchor is None:
        # Keep missing anchors visibly invalid rather than inventing
        # a coordinate for an unknown state.
        continue

    indices = group.index.tolist()

    for i, idx in enumerate(indices):
        # Tiny deterministic offsets; NOT actual district positions.
        row_offset = (i // 5) * 0.025
        col_offset = (i % 5) * 0.025

        crosswalk.loc[idx, "latitude"] = anchor[0] + row_offset
        crosswalk.loc[idx, "longitude"] = anchor[1] + col_offset
        crosswalk.loc[idx, "coordinate_source"] = "SYNTHETIC_STATE_ANCHOR_OFFSET"
        crosswalk.loc[idx, "match_notes"] = (
            "Synthetic placeholder near state anchor; not a real "
            "district coordinate. Replace before geographic analysis "
            "or district-level weather extraction."
        )

# ------------------------------------------------------------
# 6. Validate and save
# ------------------------------------------------------------
assert crosswalk["apy_geography_id"].is_unique
assert len(crosswalk) == len(apy_geo)

# Check coordinate ranges, while allowing missing coordinates for
# state labels that lack a configured anchor.
valid_coordinates = crosswalk["latitude"].notna() & crosswalk["longitude"].notna()

assert crosswalk.loc[valid_coordinates, "latitude"].between(-90, 90).all()
assert crosswalk.loc[valid_coordinates, "longitude"].between(-180, 180).all()

crosswalk.to_csv(output_path, index=False)

summary = {
    "rows": int(len(crosswalk)),
    "unique_geography_ids": int(crosswalk["apy_geography_id"].nunique()),
    "real_boundary_points": int(
        (crosswalk["coordinate_source"] == "BOUNDARY_REPRESENTATIVE_POINT").sum()
    ),
    "synthetic_placeholders": int(
        (crosswalk["coordinate_source"] == "SYNTHETIC_STATE_ANCHOR_OFFSET").sum()
    ),
    "still_missing_coordinates": int(
        crosswalk[["latitude", "longitude"]].isna().any(axis=1).sum()
    ),
    "note": (
        "Boundary points are provisional matches. Synthetic placeholders "
        "must not be used as district-specific geographic observations."
    ),
    "output_path": str(output_path),
}

with open(summary_path, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2)

print("Coordinate crosswalk created.")
print(json.dumps(summary, indent=2))
print("\nCoordinate source breakdown:")
print(crosswalk["coordinate_source"].value_counts(dropna=False))
print("\nSample rows:")
display(crosswalk.head(10))
```

```python
# ============================================================
# Fix the three missing coordinates
# Approximate representative locations, not verified centroids
# ============================================================

import pandas as pd
from pathlib import Path

project_root = Path.cwd().parent
crosswalk_path = (
    project_root / "metadata" / "apy_district_coordinate_crosswalk.csv"
)

crosswalk = pd.read_csv(crosswalk_path)

# Approximate locations for temporary geographic coverage.
# These are not official district centroids.
approximate_locations = {
    "DADRA AND NAGAR HAVELI": {
        "latitude": 20.27,
        "longitude": 73.02,
    },
    "DAMAN": {
        "latitude": 20.42,
        "longitude": 72.83,
    },
    "DIU": {
        "latitude": 20.71,
        "longitude": 70.99,
    },
}

def normalize_label(value):
    return " ".join(
        str(value).upper().replace("&", " AND ").split()
    )

target_state = normalize_label(
    "The Dadra And Nagar Haveli And Daman And Diu"
)

for district, coords in approximate_locations.items():
    mask = (
        crosswalk["State"].map(normalize_label).eq(target_state)
        & crosswalk["District"].map(normalize_label).eq(district)
    )

    if mask.sum() != 1:
        print(f"Skipping {district}: expected one matching row, found {mask.sum()}")
        continue

    crosswalk.loc[mask, "latitude"] = coords["latitude"]
    crosswalk.loc[mask, "longitude"] = coords["longitude"]
    crosswalk.loc[mask, "coordinate_source"] = "APPROXIMATE_REPRESENTATIVE_LOCATION"
    crosswalk.loc[mask, "is_synthetic"] = True
    crosswalk.loc[mask, "geography_confidence"] = "LOW_APPROXIMATE_LOCATION"
    crosswalk.loc[mask, "match_notes"] = (
        "Approximate representative location assigned manually; "
        "not a verified polygon centroid or official district HQ coordinate."
    )

crosswalk.to_csv(crosswalk_path, index=False)

# Verify the update
missing = crosswalk[
    crosswalk[["latitude", "longitude"]].isna().any(axis=1)
]

print(f"Total geography records: {len(crosswalk)}")
print(f"Missing-coordinate records: {len(missing)}")
print("\nUpdated records:")
display(
    crosswalk[
        crosswalk["coordinate_source"].eq(
            "APPROXIMATE_REPRESENTATIVE_LOCATION"
        )
    ][
        [
            "State", "District", "latitude", "longitude",
            "coordinate_source", "geography_confidence"
        ]
    ]
)
```

```python
# ============================================================
# NASA POWER: 10-location daily weather pilot, 2013–2024
# Variables:
#   PRECTOTCORR — corrected precipitation (mm/day)
#   T2M_MAX     — daily maximum temperature (°C)
#   T2M_MIN     — daily minimum temperature (°C)
# ============================================================

import pandas as pd
import requests
import time
import json
from pathlib import Path
from datetime import datetime

project_root = Path.cwd().parent
crosswalk_path = project_root / "metadata" / "apy_district_coordinate_crosswalk.csv"
output_dir = project_root / "data" / "raw" / "weather_pilot_2013_2024"
manifest_path = project_root / "reports" / "data_audit" / "nasa_power_pilot_manifest.csv"

output_dir.mkdir(parents=True, exist_ok=True)
manifest_path.parent.mkdir(parents=True, exist_ok=True)

# Rebuild the same 10-location pilot from the saved crosswalk.
crosswalk = pd.read_csv(crosswalk_path)

pilot_states = [
    "Punjab", "Andhra Pradesh", "Madhya Pradesh", "Assam", "Tamil Nadu"
]

pilot = (
    crosswalk[
        crosswalk["coordinate_source"].eq("BOUNDARY_REPRESENTATIVE_POINT")
        & crosswalk["latitude"].notna()
        & crosswalk["longitude"].notna()
        & crosswalk["State"].str.casefold().isin(
            [s.casefold() for s in pilot_states]
        )
    ]
    .sort_values(["State", "District"])
    .groupby("State", group_keys=False)
    .head(2)
    .reset_index(drop=True)
)

assert len(pilot) == 10, f"Expected 10 pilot locations, found {len(pilot)}"

base_url = "https://power.larc.nasa.gov/api/temporal/daily/point"
parameters = "PRECTOTCORR,T2M_MAX,T2M_MIN"
session = requests.Session()

manifest = []

for _, location in pilot.iterrows():
    geo_id = location["apy_geography_id"]
    state = location["State"]
    district = location["District"]

    # Use a stable filename based on the geography ID.
    output_file = output_dir / f"{geo_id}_weather_2013_2024.csv"

    # Skip an existing file only if it parses and has data.
    if output_file.exists():
        try:
            existing = pd.read_csv(output_file, comment="#")
            if not existing.empty:
                manifest.append({
                    "apy_geography_id": geo_id,
                    "State": state,
                    "District": district,
                    "status": "SKIPPED_EXISTING_FILE",
                    "file_path": str(output_file),
                    "rows": len(existing),
                    "error": "",
                })
                continue
        except Exception:
            pass  # Retry download if the existing file is unreadable.

    params = {
        "parameters": parameters,
        "community": "AG",
        "longitude": float(location["longitude"]),
        "latitude": float(location["latitude"]),
        "start": "20130101",
        "end": "20241231",
        "format": "CSV",
        "time-standard": "UTC",
    }

    try:
        response = session.get(base_url, params=params, timeout=90)
        response.raise_for_status()

        # NASA POWER CSV responses may include metadata comment lines.
        output_file.write_text(response.text, encoding="utf-8")

        # Validate that the file is parseable and non-empty.
        downloaded = pd.read_csv(output_file, comment="#")

        if downloaded.empty:
            raise ValueError("Downloaded CSV contains no data rows.")

        manifest.append({
            "apy_geography_id": geo_id,
            "State": state,
            "District": district,
            "status": "DOWNLOADED",
            "file_path": str(output_file),
            "rows": len(downloaded),
            "error": "",
        })

        print(f"Downloaded: {state} — {district} ({len(downloaded)} rows)")

    except Exception as exc:
        manifest.append({
            "apy_geography_id": geo_id,
            "State": state,
            "District": district,
            "status": "FAILED",
            "file_path": str(output_file),
            "rows": 0,
            "error": str(exc),
        })
        print(f"FAILED: {state} — {district}: {exc}")

    # Be polite to the API; do not make parallel requests.
    time.sleep(1.5)

    # Save progress after every location so interruptions are recoverable.
    pd.DataFrame(manifest).to_csv(manifest_path, index=False)

session.close()

manifest_df = pd.DataFrame(manifest)
manifest_df.to_csv(manifest_path, index=False)

print("\n--- Pilot acquisition summary ---")
print(manifest_df["status"].value_counts(dropna=False))
print(f"Manifest: {manifest_path}")
print(f"CSV directory: {output_dir}")
```

```python

# Cell: Parse and validate existing NASA POWER weather files
# No network requests are made in this cell.

from pathlib import Path
import pandas as pd
import numpy as np
import calendar

project_root = Path.cwd().parent

input_dir = project_root / "data" / "raw" / "weather_pilot_2013_2024"
output_dir = project_root / "data" / "processed" / "weather_pilot_2013_2024"
report_dir = project_root / "reports" / "data_audit"

output_dir.mkdir(parents=True, exist_ok=True)
report_dir.mkdir(parents=True, exist_ok=True)

required_columns = {
    "YEAR", "DOY", "PRECTOTCORR", "T2M_MAX", "T2M_MIN"
}

weather_frames = []
audit_rows = []

files = sorted(input_dir.glob("*.csv"))
print(f"Files found: {len(files)}")

for file_path in files:
    try:
        # Locate the actual table header instead of parsing NASA's metadata.
        with open(file_path, "r", encoding="utf-8-sig") as file:
            lines = file.readlines()

        header_idx = None

        for i, line in enumerate(lines):
            columns = {
                value.strip().upper()
                for value in line.strip().split(",")
            }

            if required_columns.issubset(columns):
                header_idx = i
                break

        if header_idx is None:
            raise ValueError("Could not locate the expected CSV column header.")

        # Read only the table section, skipping the metadata preamble.
        df = pd.read_csv(
            file_path,
            skiprows=header_idx,
            skip_blank_lines=True
        )

        df.columns = df.columns.str.strip().str.upper()

        missing_columns = required_columns - set(df.columns)
        if missing_columns:
            raise ValueError(f"Missing columns: {sorted(missing_columns)}")

        # Convert data columns to numeric values.
        numeric_columns = [
            "YEAR", "DOY", "PRECTOTCORR", "T2M_MAX", "T2M_MIN"
        ]
        for col in numeric_columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

        if df[["YEAR", "DOY"]].isna().any().any():
            raise ValueError("Invalid YEAR or DOY values found.")

        # Build dates from year and day-of-year.
        df["DATE"] = (
            pd.to_datetime(df["YEAR"].astype(int).astype(str), format="%Y")
            + pd.to_timedelta(df["DOY"].astype(int) - 1, unit="D")
        )

        # Reject impossible day-of-year values that roll into another year.
        invalid_doy = (
            (df["DOY"] < 1)
            | (
                df["DOY"]
                > df["YEAR"].apply(
                    lambda year: 366 if calendar.isleap(int(year)) else 365
                )
            )
        )
        if invalid_doy.any():
            raise ValueError("Invalid day-of-year values found.")

        # Derive the geography ID from the filename.
        geography_id = file_path.name.split("_weather_")[0]
        df["geography_id"] = geography_id
        df["source_file"] = file_path.name

        # NASA POWER uses -999 as its missing-data sentinel.
        weather_vars = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]
        df[weather_vars] = df[weather_vars].replace(-999, np.nan)

        # Check the requested period and expected daily calendar.
        expected_dates = pd.date_range("2013-01-01", "2024-12-31", freq="D")
        actual_dates = pd.DatetimeIndex(df["DATE"].sort_values())

        duplicate_dates = int(df["DATE"].duplicated().sum())
        missing_dates = len(expected_dates.difference(actual_dates))
        unexpected_dates = len(actual_dates.difference(expected_dates))

        # Record data-quality findings; do not silently correct anomalies.
        audit_rows.append({
            "geography_id": geography_id,
            "source_file": file_path.name,
            "status": "PARSED",
            "row_count": len(df),
            "expected_rows": len(expected_dates),
            "duplicate_dates": duplicate_dates,
            "missing_dates": missing_dates,
            "unexpected_dates": unexpected_dates,
            "missing_precipitation": int(df["PRECTOTCORR"].isna().sum()),
            "missing_tmax": int(df["T2M_MAX"].isna().sum()),
            "missing_tmin": int(df["T2M_MIN"].isna().sum()),
            "negative_precipitation": int(
                (df["PRECTOTCORR"] < 0).sum()
            ),
            "tmin_greater_than_tmax": int(
                (df["T2M_MIN"] > df["T2M_MAX"]).sum()
            ),
            "date_start": df["DATE"].min(),
            "date_end": df["DATE"].max(),
            "error": ""
        })

        weather_frames.append(df)

    except Exception as exc:
        audit_rows.append({
            "geography_id": file_path.name.split("_weather_")[0],
            "source_file": file_path.name,
            "status": "PARSE_FAILED",
            "error": str(exc)
        })

# Save parsed data and audit results.
audit_df = pd.DataFrame(audit_rows)
audit_path = report_dir / "nasa_power_pilot_parse_audit.csv"
audit_df.to_csv(audit_path, index=False)

if weather_frames:
    weather_df = pd.concat(weather_frames, ignore_index=True)
    weather_df = weather_df.sort_values(
        ["geography_id", "DATE"]
    ).reset_index(drop=True)

    data_path = output_dir / "nasa_power_weather_pilot_2013_2024.csv"
    weather_df.to_csv(data_path, index=False)

    print("\nParsing summary")
    print("----------------")
    print(f"Successfully parsed files: {len(weather_frames)} / {len(files)}")
    print(f"Combined rows: {len(weather_df):,}")
    print(f"Expected rows if all 10 files are complete: {len(files) * 4383:,}")
    print(f"Combined data: {data_path}")
    print(f"Audit report: {audit_path}")

    display(audit_df)
else:
    print("No files were parsed successfully. Inspect the audit report.")
    display(audit_df)
```

```python

# Cell: Deeper plausibility checks for the NASA POWER weather pilot

from pathlib import Path
import pandas as pd
import numpy as np

project_root = Path.cwd().parent

data_path = (
    project_root / "data" / "processed"
    / "weather_pilot_2013_2024"
    / "nasa_power_weather_pilot_2013_2024.csv"
)

df = pd.read_csv(data_path, parse_dates=["DATE"])

weather_vars = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]

print("DATASET OVERVIEW")
print("=" * 60)
print(f"Shape: {df.shape}")
print(f"Unique locations: {df['geography_id'].nunique()}")
print(f"Date range: {df['DATE'].min().date()} to {df['DATE'].max().date()}")
print(f"Missing values:\n{df[weather_vars].isna().sum()}")

# 1. Distribution and range checks.
print("\nWEATHER VARIABLE SUMMARY")
display(
    df[weather_vars].describe(
        percentiles=[0.01, 0.05, 0.50, 0.95, 0.99]
    ).T
)

# 2. Flag extreme values for inspection, not automatic deletion.
# These are broad screening thresholds, not official validity limits.
extreme_flags = {
    "precipitation_below_zero": df["PRECTOTCORR"] < 0,
    "precipitation_above_500_mm_day": df["PRECTOTCORR"] > 500,
    "tmax_below_minus_10C": df["T2M_MAX"] < -10,
    "tmax_above_60C": df["T2M_MAX"] > 60,
    "tmin_below_minus_15C": df["T2M_MIN"] < -15,
    "tmin_above_45C": df["T2M_MIN"] > 45,
    "tmin_greater_than_tmax": df["T2M_MIN"] > df["T2M_MAX"],
}

print("\nEXTREME-VALUE SCREEN")
extreme_summary = pd.DataFrame([
    {"check": name, "flagged_rows": int(mask.sum())}
    for name, mask in extreme_flags.items()
])
display(extreme_summary)

# 3. Per-location summary to detect suspiciously identical or unusual series.
location_summary = (
    df.groupby("geography_id")
    .agg(
        days=("DATE", "count"),
        precipitation_mean=("PRECTOTCORR", "mean"),
        precipitation_max=("PRECTOTCORR", "max"),
        tmax_mean=("T2M_MAX", "mean"),
        tmin_mean=("T2M_MIN", "mean"),
        tmax_min=("T2M_MAX", "min"),
        tmax_max=("T2M_MAX", "max"),
        tmin_min=("T2M_MIN", "min"),
        tmin_max=("T2M_MIN", "max"),
    )
    .reset_index()
)

print("\nPER-LOCATION SUMMARY")
display(location_summary.round(2))

# 4. Check whether any locations have identical daily weather series.
wide = df.pivot(index="DATE", columns="geography_id", values="T2M_MAX")
correlations = wide.corr()

pairs = []
location_ids = list(correlations.columns)

for i in range(len(location_ids)):
    for j in range(i + 1, len(location_ids)):
        a, b = location_ids[i], location_ids[j]
        pairs.append({
            "location_1": a,
            "location_2": b,
            "tmax_correlation": correlations.loc[a, b],
            "identical_tmax_series": bool(
                np.allclose(
                    wide[a].to_numpy(),
                    wide[b].to_numpy(),
                    equal_nan=True
                )
            )
        })

pair_checks = pd.DataFrame(pairs).sort_values(
    "tmax_correlation", ascending=False
)

print("\nMOST SIMILAR LOCATION PAIRS")
display(pair_checks.head(10).round(4))
```

```python

# Cell: Investigate identical weather series and coordinate mapping

from pathlib import Path
import pandas as pd
import numpy as np
import re

project_root = Path.cwd().parent

weather_dir = (
    project_root / "data" / "raw" / "weather_pilot_2013_2024"
)
weather_path = (
    project_root / "data" / "processed"
    / "weather_pilot_2013_2024"
    / "nasa_power_weather_pilot_2013_2024.csv"
)
crosswalk_path = (
    project_root / "metadata" / "apy_district_coordinate_crosswalk.csv"
)

weather = pd.read_csv(weather_path, parse_dates=["DATE"])
crosswalk = pd.read_csv(crosswalk_path)

pair_ids = ["APY-GEO-0055", "APY-GEO-0056"]
variables = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]

# 1. Inspect crosswalk records.
print("CROSSWALK COORDINATES")
print("=" * 60)

display(
    crosswalk.loc[
        crosswalk["apy_geography_id"].isin(pair_ids),
        [
            "apy_geography_id", "State", "District",
            "latitude", "longitude", "coordinate_source",
            "is_synthetic", "geography_confidence",
            "matched_boundary_name", "match_notes"
        ]
    ]
)

# 2. Compare each weather variable for the two locations.
print("\nPAIRWISE WEATHER COMPARISON")
print("=" * 60)

for variable in variables:
    left = (
        weather.loc[weather["geography_id"] == pair_ids[0]]
        .sort_values("DATE")[variable]
        .to_numpy()
    )
    right = (
        weather.loc[weather["geography_id"] == pair_ids[1]]
        .sort_values("DATE")[variable]
        .to_numpy()
    )

    if len(left) != len(right):
        print(f"{variable}: row counts differ ({len(left)} vs {len(right)})")
        continue

    print(
        f"{variable}: identical={np.array_equal(left, right, equal_nan=True)}, "
        f"max_absolute_difference={np.nanmax(np.abs(left - right)):.6f}"
    )

# 3. Read coordinates reported by NASA POWER in the original downloaded files.
print("\nORIGINAL NASA POWER FILE METADATA")
print("=" * 60)

for geography_id in pair_ids:
    matches = list(
        weather_dir.glob(f"{geography_id}_weather_*.csv")
    )

    if not matches:
        print(f"{geography_id}: original file not found")
        continue

    source_text = matches[0].read_text(encoding="utf-8-sig")

    latitude_match = re.search(
        r"Location:\s*latitude\s+([-+]?\d+(?:\.\d+)?)",
        source_text
    )
    longitude_match = re.search(
        r"longitude\s+([-+]?\d+(?:\.\d+)?)",
        source_text
    )

    print(f"\n{geography_id}:")
    print(
        "Latitude:",
        latitude_match.group(1) if latitude_match else "not found"
    )
    print(
        "Longitude:",
        longitude_match.group(1) if longitude_match else "not found"
    )

# 4. Calculate the approximate distance between crosswalk coordinates.
coords = crosswalk.set_index("apy_geography_id")

if all(pid in coords.index for pid in pair_ids):
    a = coords.loc[pair_ids[0]]
    b = coords.loc[pair_ids[1]]

    if all(pd.notna(v) for v in [
        a["latitude"], a["longitude"],
        b["latitude"], b["longitude"]
    ]):
        lat1, lon1 = np.radians([a["latitude"], a["longitude"]])
        lat2, lon2 = np.radians([b["latitude"], b["longitude"]])

        haversine = 2 * 6371 * np.arcsin(np.sqrt(
            np.sin((lat2 - lat1) / 2) ** 2
            + np.cos(lat1) * np.cos(lat2)
            * np.sin((lon2 - lon1) / 2) ** 2
        ))

        print(f"\nCrosswalk coordinate distance: {haversine:.2f} km")
    else:
        print("\nCannot calculate distance: coordinates are missing.")
else:
    print("\nOne or both geography IDs are absent from the crosswalk.")
```

```python

# STEP 3: Create a weather acquisition eligibility manifest
# This classifies geography records; it does NOT start API downloads.

from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent

crosswalk_path = (
    project_root / "metadata" / "apy_district_coordinate_crosswalk.csv"
)

crosswalk = pd.read_csv(crosswalk_path)

# Treat only boundary-derived representative points as provisional candidates.
# Synthetic and approximate coordinates are excluded from district-level requests.
crosswalk["acquisition_eligible"] = (
    crosswalk["coordinate_source"].eq("BOUNDARY_REPRESENTATIVE_POINT")
    & ~crosswalk["is_synthetic"].astype(str).str.lower().eq("true")
    & crosswalk["latitude"].between(-90, 90)
    & crosswalk["longitude"].between(-180, 180)
)

crosswalk["acquisition_status"] = crosswalk["acquisition_eligible"].map(
    {
        True: "ELIGIBLE_FOR_GRID_AUDIT",
        False: "HOLD_COORDINATE_REVIEW"
    }
)

# Keep the geographic uncertainty explicit.
crosswalk["weather_interpretation"] = (
    "Gridded weather estimate near a provisional district representative point; "
    "not a district-wide observation"
)

manifest_columns = [
    "apy_geography_id",
    "State",
    "District",
    "latitude",
    "longitude",
    "coordinate_source",
    "is_synthetic",
    "geography_confidence",
    "acquisition_eligible",
    "acquisition_status",
    "weather_interpretation"
]

manifest = crosswalk[manifest_columns].copy()

print("ACQUISITION ELIGIBILITY")
print("-" * 60)
display(manifest["acquisition_status"].value_counts().to_frame())

print("\nELIGIBILITY BY COORDINATE SOURCE")
display(
    manifest.groupby(
        ["coordinate_source", "acquisition_status"],
        dropna=False
    ).size().rename("geography_count").reset_index()
)

# Save the manifest separately; original crosswalk remains unchanged.
output_path = (
    project_root / "metadata" / "weather_acquisition_manifest.csv"
)
output_path.parent.mkdir(parents=True, exist_ok=True)
manifest.to_csv(output_path, index=False)

print("\nManifest saved to:", output_path)
print("Total records:", len(manifest))
print("Eligible provisional points:", int(manifest["acquisition_eligible"].sum()))
print("Held for coordinate review:", int((~manifest["acquisition_eligible"]).sum()))
```

```python

# STEP 4: Screen nearby pilot locations before expanding acquisition
# This is a distance-based screening step, NOT an exact grid-cell mapping.

from pathlib import Path
import pandas as pd
import numpy as np
from itertools import combinations

project_root = Path.cwd().parent

crosswalk = pd.read_csv(
    project_root / "metadata" / "weather_acquisition_manifest.csv"
)

eligible = crosswalk.loc[
    crosswalk["acquisition_status"].eq("ELIGIBLE_FOR_GRID_AUDIT")
].copy()

# Haversine distance between two latitude/longitude points, in kilometres.
def haversine_km(lat1, lon1, lat2, lon2):
    earth_radius_km = 6371.0088

    lat1, lon1, lat2, lon2 = np.radians(
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )

    return 2 * earth_radius_km * np.arcsin(np.sqrt(a))

# Screen all eligible pairs within 60 km.
# This threshold is for triage only, not a definition of grid-cell membership.
records = eligible[
    ["apy_geography_id", "State", "District", "latitude", "longitude"]
].to_dict("records")

nearby_pairs = []

for a, b in combinations(records, 2):
    distance = haversine_km(
        a["latitude"], a["longitude"],
        b["latitude"], b["longitude"]
    )

    if distance <= 60:
        nearby_pairs.append({
            "geography_id_1": a["apy_geography_id"],
            "State_1": a["State"],
            "District_1": a["District"],
            "latitude_1": a["latitude"],
            "longitude_1": a["longitude"],
            "geography_id_2": b["apy_geography_id"],
            "State_2": b["State"],
            "District_2": b["District"],
            "latitude_2": b["latitude"],
            "longitude_2": b["longitude"],
            "distance_km": round(distance, 2)
        })

nearby_df = pd.DataFrame(nearby_pairs)

print("Eligible provisional coordinates:", len(eligible))
print("Pairs within 60 km:", len(nearby_df))

if not nearby_df.empty:
    nearby_df = nearby_df.sort_values("distance_km")
    display(nearby_df.head(30))

    output_path = (
        project_root / "reports" / "data_audit"
        / "weather_nearby_coordinate_pairs.csv"
    )
    output_path.parent.mkdir(parents=True, exist_ok=True)
    nearby_df.to_csv(output_path, index=False)
    print("Nearby-pair report saved to:", output_path)
else:
    print("No pairs within the screening threshold.")
```

```python

# STEP 7: Match the Assam pilot locations to explicit regional grid points
# Small, read-only API test: 1 parameter, 3 days.

import requests
import pandas as pd
import numpy as np
from pathlib import Path

project_root = Path.cwd().parent

regional_url = "https://power.larc.nasa.gov/api/temporal/daily/regional"

params = {
    "parameters": "T2M_MAX",
    "community": "AG",
    "longitude-min": 90.5,
    "longitude-max": 92.5,
    "latitude-min": 25.5,
    "latitude-max": 27.5,
    "start": "20130101",
    "end": "20130103",
    "format": "JSON",
    "time-standard": "UTC"
}

response = requests.get(regional_url, params=params, timeout=60)
response.raise_for_status()
payload = response.json()

# Extract explicit regional grid-point coordinates and temperature series.
grid_rows = []

for feature in payload["features"]:
    longitude, latitude, *elevation = feature["geometry"]["coordinates"]
    values = feature["properties"]["parameter"]["T2M_MAX"]

    grid_rows.append({
        "grid_latitude": latitude,
        "grid_longitude": longitude,
        "grid_elevation_m": elevation[0] if elevation else np.nan,
        "T2M_MAX_20130101": values.get("20130101"),
        "T2M_MAX_20130102": values.get("20130102"),
        "T2M_MAX_20130103": values.get("20130103")
    })

grid = pd.DataFrame(grid_rows)

# Find the nearest regional grid point to each pilot coordinate.
pilot_points = [
    {
        "geography_id": "APY-GEO-0055",
        "District": "Baksa",
        "latitude": 26.632399,
        "longitude": 91.335490
    },
    {
        "geography_id": "APY-GEO-0056",
        "District": "Barpeta",
        "latitude": 26.358313,
        "longitude": 91.058334
    }
]

def haversine_km(lat1, lon1, lat2, lon2):
    radius = 6371.0088
    lat1, lon1, lat2, lon2 = np.radians(
        [lat1, lon1, lat2, lon2]
    )
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
    )
    return 2 * radius * np.arcsin(np.sqrt(a))

matches = []

for point in pilot_points:
    distances = grid.apply(
        lambda row: haversine_km(
            point["latitude"], point["longitude"],
            row["grid_latitude"], row["grid_longitude"]
        ),
        axis=1
    )

    nearest_index = distances.idxmin()
    nearest = grid.loc[nearest_index]

    matches.append({
        **point,
        "nearest_grid_latitude": nearest["grid_latitude"],
        "nearest_grid_longitude": nearest["grid_longitude"],
        "distance_to_grid_point_km": round(distances.loc[nearest_index], 3),
        "grid_elevation_m": nearest["grid_elevation_m"],
        "regional_tmax_20130101": nearest["T2M_MAX_20130101"],
        "regional_tmax_20130102": nearest["T2M_MAX_20130102"],
        "regional_tmax_20130103": nearest["T2M_MAX_20130103"]
    })

matches_df = pd.DataFrame(matches)

print("NEAREST REGIONAL GRID POINTS")
display(matches_df)

print("\nGRID-POINT MATCH")
same_grid_point = (
    matches_df.loc[0, "nearest_grid_latitude"]
    == matches_df.loc[1, "nearest_grid_latitude"]
    and
    matches_df.loc[0, "nearest_grid_longitude"]
    == matches_df.loc[1, "nearest_grid_longitude"]
)
print("Both locations map to the same nearest regional grid point:", same_grid_point)

# Compare regional values with the existing point pilot for these dates.
weather_path = (
    project_root / "data" / "processed" / "weather_pilot_2013_2024"
    / "nasa_power_weather_pilot_2013_2024.csv"
)

weather = pd.read_csv(weather_path)
weather["DATE"] = pd.to_datetime(weather["DATE"])

pilot_comparison = weather[
    weather["geography_id"].isin(
        ["APY-GEO-0055", "APY-GEO-0056"]
    )
    & weather["DATE"].isin(pd.to_datetime([
        "2013-01-01", "2013-01-02", "2013-01-03"
    ]))
][["geography_id", "DATE", "T2M_MAX"]]

print("\nEXISTING POINT-API VALUES")
display(pilot_comparison)
```

```python

# STEP 8: Plan regional grid coverage for all eligible provisional points

from pathlib import Path
import pandas as pd
import math

project_root = Path.cwd().parent

manifest = pd.read_csv(
    project_root / "metadata" / "weather_acquisition_manifest.csv"
)

eligible = manifest[
    manifest["acquisition_status"].eq("ELIGIBLE_FOR_GRID_AUDIT")
].copy()

lat_min = float(eligible["latitude"].min())
lat_max = float(eligible["latitude"].max())
lon_min = float(eligible["longitude"].min())
lon_max = float(eligible["longitude"].max())

# Add a small margin around the candidate coordinates.
margin = 0.25

bounds = {
    "latitude_min": max(-90, lat_min - margin),
    "latitude_max": min(90, lat_max + margin),
    "longitude_min": max(-180, lon_min - margin),
    "longitude_max": min(180, lon_max + margin),
}

lat_span = bounds["latitude_max"] - bounds["latitude_min"]
lon_span = bounds["longitude_max"] - bounds["longitude_min"]

# Regional endpoint allows at least 2 degrees and documentation indicates
# a maximum processed region of about 4.5 x 4.5 degrees per request.
tile_size = 4.5

lat_tiles = math.ceil(lat_span / tile_size)
lon_tiles = math.ceil(lon_span / tile_size)

print("ELIGIBLE COORDINATE EXTENT")
print("-" * 60)
print(f"Latitude:  {lat_min:.4f} to {lat_max:.4f}")
print(f"Longitude: {lon_min:.4f} to {lon_max:.4f}")
print(f"Latitude span with margin:  {lat_span:.4f} degrees")
print(f"Longitude span with margin: {lon_span:.4f} degrees")
print("\nBounding box:", bounds)
print("\nApproximate 4.5-degree tile count:")
print("Latitude tiles:", lat_tiles)
print("Longitude tiles:", lon_tiles)
print("Total tiles (rough estimate):", lat_tiles * lon_tiles)

print(
    "\nNote: this is only a planning estimate. "
    "A nationwide bounding box cannot be queried as one regional request, "
    "and each tile must satisfy the API's minimum extent."
)
```

```python

# STEP 9: Build an overlapping regional tile plan and check coverage
# Planning only: no network requests are made.

import pandas as pd
from pathlib import Path
import math

project_root = Path.cwd().parent

manifest = pd.read_csv(
    project_root / "metadata" / "weather_acquisition_manifest.csv"
)

eligible = manifest[
    manifest["acquisition_status"].eq("ELIGIBLE_FOR_GRID_AUDIT")
].copy()

# Slightly padded coordinate extent.
margin = 0.25
lat_min = eligible["latitude"].min() - margin
lat_max = eligible["latitude"].max() + margin
lon_min = eligible["longitude"].min() - margin
lon_max = eligible["longitude"].max() + margin

# Regional API maximum area is approximately 4.5 x 4.5 degrees.
# Use 4.0-degree tiles so neighboring tiles overlap.
tile_size = 4.0

def make_edges(minimum, maximum, size):
    edges = [minimum]
    while edges[-1] < maximum:
        edges.append(min(edges[-1] + size, maximum))
    return edges

lat_edges = make_edges(lat_min, lat_max, tile_size)
lon_edges = make_edges(lon_min, lon_max, tile_size)

tiles = []
tile_id = 1

for i in range(len(lat_edges) - 1):
    for j in range(len(lon_edges) - 1):
        tiles.append({
            "tile_id": f"T{tile_id:03d}",
            "latitude_min": lat_edges[i],
            "latitude_max": lat_edges[i + 1],
            "longitude_min": lon_edges[j],
            "longitude_max": lon_edges[j + 1],
        })
        tile_id += 1

tiles_df = pd.DataFrame(tiles)

# Confirm that every eligible point is contained in at least one tile.
coverage_counts = []

for _, point in eligible.iterrows():
    covered = (
        (tiles_df["latitude_min"] <= point["latitude"])
        & (tiles_df["latitude_max"] >= point["latitude"])
        & (tiles_df["longitude_min"] <= point["longitude"])
        & (tiles_df["longitude_max"] >= point["longitude"])
    ).sum()

    coverage_counts.append({
        "apy_geography_id": point["apy_geography_id"],
        "latitude": point["latitude"],
        "longitude": point["longitude"],
        "covering_tile_count": int(covered)
    })

coverage_df = pd.DataFrame(coverage_counts)

print("TILE PLAN")
print("-" * 60)
print("Number of tiles:", len(tiles_df))
print("Latitude tile intervals:", len(lat_edges) - 1)
print("Longitude tile intervals:", len(lon_edges) - 1)
print("Tile dimensions: at most 4.0° × 4.0°")
print("\nEligible points:", len(eligible))
print("Points covered by at least one tile:",
      int((coverage_df["covering_tile_count"] >= 1).sum()))
print("Points with no tile coverage:",
      int((coverage_df["covering_tile_count"] == 0).sum()))

print("\nTILE DIMENSION CHECK")
print("Largest latitude span:",
      round((tiles_df["latitude_max"] - tiles_df["latitude_min"]).max(), 3))
print("Largest longitude span:",
      round((tiles_df["longitude_max"] - tiles_df["longitude_min"]).max(), 3))

display(tiles_df.head(10))
display(coverage_df["covering_tile_count"].value_counts().sort_index().rename_axis(
    "number_of_covering_tiles"
).to_frame("number_of_points"))
```

```python

# STEP 11: Retrieve a short regional grid sample for all planned tiles
# One variable, three days. This is an audit, not the full weather download.

import requests
import pandas as pd
import numpy as np
import time
from pathlib import Path

project_root = Path.cwd().parent
regional_url = "https://power.larc.nasa.gov/api/temporal/daily/regional"

grid_records = []
request_log = []

for _, tile in tiles_df.iterrows():
    tile_id = tile["tile_id"]

    params = {
        "parameters": "T2M_MAX",
        "community": "AG",
        "longitude-min": round(float(tile["longitude_min"]), 4),
        "longitude-max": round(float(tile["longitude_max"]), 4),
        "latitude-min": round(float(tile["latitude_min"]), 4),
        "latitude-max": round(float(tile["latitude_max"]), 4),
        "start": "20130101",
        "end": "20130103",
        "format": "JSON",
        "time-standard": "UTC"
    }

    status = None
    error_message = None
    returned_count = 0

    # Retry transient failures up to three times.
    for attempt in range(1, 4):
        try:
            response = requests.get(
                regional_url, params=params, timeout=90
            )
            status = response.status_code
            response.raise_for_status()
            payload = response.json()

            features = payload.get("features", [])
            if not features:
                raise ValueError("API returned no grid features")

            for feature in features:
                longitude, latitude, *elevation = (
                    feature["geometry"]["coordinates"]
                )
                values = feature["properties"]["parameter"]["T2M_MAX"]

                grid_records.append({
                    "tile_id": tile_id,
                    "grid_latitude": float(latitude),
                    "grid_longitude": float(longitude),
                    "grid_elevation_m": (
                        float(elevation[0]) if elevation else np.nan
                    ),
                    "T2M_MAX_20130101": values.get("20130101"),
                    "T2M_MAX_20130102": values.get("20130102"),
                    "T2M_MAX_20130103": values.get("20130103")
                })

            returned_count = len(features)
            error_message = None
            break

        except Exception as exc:
            error_message = f"{type(exc).__name__}: {exc}"
            if attempt < 3:
                time.sleep(2 * attempt)

    request_log.append({
        "tile_id": tile_id,
        "http_status": status,
        "attempts_used": attempt,
        "grid_points_returned": returned_count,
        "error": error_message
    })

    print(
        f"{tile_id}: status={status}, "
        f"points={returned_count}, attempts={attempt}"
    )

    # Be considerate of the public API.
    time.sleep(0.3)

grid_raw = pd.DataFrame(grid_records)
request_audit = pd.DataFrame(request_log)

audit_dir = project_root / "reports" / "data_audit"
audit_dir.mkdir(parents=True, exist_ok=True)

grid_raw_path = audit_dir / "nasa_power_regional_grid_raw_sample.csv"
request_log_path = audit_dir / "nasa_power_regional_grid_request_audit.csv"

grid_raw.to_csv(grid_raw_path, index=False)
request_audit.to_csv(request_log_path, index=False)

# Check duplicate coordinates caused by tile-edge overlap.
coord_cols = ["grid_latitude", "grid_longitude"]
unique_coords = grid_raw[coord_cols].drop_duplicates()

duplicate_rows = grid_raw[
    grid_raw.duplicated(coord_cols, keep=False)
].sort_values(coord_cols)

# For repeated coordinates, check whether the sample values agree.
value_cols = [
    "T2M_MAX_20130101",
    "T2M_MAX_20130102",
    "T2M_MAX_20130103"
]

conflicts = []
for coords, group in duplicate_rows.groupby(coord_cols, dropna=False):
    if (group[value_cols].nunique(dropna=False) > 1).any():
        conflicts.append({
            "grid_latitude": coords[0],
            "grid_longitude": coords[1],
            "tiles": ", ".join(sorted(group["tile_id"].unique())),
            "distinct_values_per_day": group[value_cols].nunique(
                dropna=False
            ).to_dict()
        })

print("\nGRID INVENTORY SUMMARY")
print("-" * 60)
print("Tiles planned:", len(tiles_df))
print("Tiles with request errors:", int(request_audit["error"].notna().sum()))
print("Raw grid records:", len(grid_raw))
print("Unique grid coordinates:", len(unique_coords))
print("Duplicate records from overlapping tile responses:",
      len(grid_raw) - len(unique_coords))
print("Coordinates with conflicting sample values:", len(conflicts))
print("Raw grid inventory saved:", grid_raw_path)
print("Request audit saved:", request_log_path)

if request_audit["error"].notna().any():
    print("\nFAILED REQUESTS")
    display(request_audit[request_audit["error"].notna()])

if conflicts:
    print("\nCONFLICTING OVERLAP VALUES")
    display(pd.DataFrame(conflicts))

display(request_audit)
```

```python

# STEP 13: Repair regional tiles to satisfy NASA POWER minimum bounds
# Planning only: no API requests are made.

import pandas as pd
import math

# Reuse the padded extent and eligible coordinates from earlier cells.
lat_min = float(eligible["latitude"].min()) - 0.25
lat_max = float(eligible["latitude"].max()) + 0.25
lon_min = float(eligible["longitude"].min()) - 0.25
lon_max = float(eligible["longitude"].max()) + 0.25

# Use 4-degree intervals, but ensure the final interval is at least 2 degrees.
tile_size = 4.0

def make_valid_edges(minimum, maximum, size, min_final_span=2.0):
    edges = [minimum]

    while maximum - edges[-1] > size:
        next_edge = edges[-1] + size

        # Avoid creating a final sliver smaller than the API minimum.
        if maximum - next_edge < min_final_span:
            next_edge = maximum - min_final_span

        edges.append(next_edge)

    edges.append(maximum)
    return edges

lat_edges = make_valid_edges(lat_min, lat_max, tile_size)
lon_edges = make_valid_edges(lon_min, lon_max, tile_size)

repaired_tiles = []
tile_id = 1

for i in range(len(lat_edges) - 1):
    for j in range(len(lon_edges) - 1):
        repaired_tiles.append({
            "tile_id": f"R{tile_id:03d}",
            "latitude_min": lat_edges[i],
            "latitude_max": lat_edges[i + 1],
            "longitude_min": lon_edges[j],
            "longitude_max": lon_edges[j + 1],
        })
        tile_id += 1

repaired_tiles_df = pd.DataFrame(repaired_tiles)

repaired_tiles_df["latitude_span"] = (
    repaired_tiles_df["latitude_max"]
    - repaired_tiles_df["latitude_min"]
)
repaired_tiles_df["longitude_span"] = (
    repaired_tiles_df["longitude_max"]
    - repaired_tiles_df["longitude_min"]
)

# Coverage check for all eligible coordinates.
coverage_counts = []
for _, point in eligible.iterrows():
    covered = (
        (repaired_tiles_df["latitude_min"] <= point["latitude"])
        & (repaired_tiles_df["latitude_max"] >= point["latitude"])
        & (repaired_tiles_df["longitude_min"] <= point["longitude"])
        & (repaired_tiles_df["longitude_max"] >= point["longitude"])
    ).sum()

    coverage_counts.append(int(covered))

print("Repaired tile count:", len(repaired_tiles_df))
print("Tiles below 2-degree latitude minimum:",
      int((repaired_tiles_df["latitude_span"] < 2.0).sum()))
print("Tiles below 2-degree longitude minimum:",
      int((repaired_tiles_df["longitude_span"] < 2.0).sum()))
print("Tiles exceeding 4.5-degree latitude span:",
      int((repaired_tiles_df["latitude_span"] > 4.5).sum()))
print("Tiles exceeding 4.5-degree longitude span:",
      int((repaired_tiles_df["longitude_span"] > 4.5).sum()))
print("Eligible points uncovered:",
      sum(count == 0 for count in coverage_counts))
print("Eligible points covered by multiple tiles:",
      sum(count > 1 for count in coverage_counts))

display(repaired_tiles_df.tail(12))
```

```python

# STEP 15: Retrieve grid inventory using the repaired tile plan
# One variable, three days per request. Saves a new audit version.

import requests
import pandas as pd
import numpy as np
import time
from pathlib import Path

project_root = Path.cwd().parent
audit_dir = project_root / "reports" / "data_audit"
audit_dir.mkdir(parents=True, exist_ok=True)

regional_url = "https://power.larc.nasa.gov/api/temporal/daily/regional"

grid_records = []
request_log = []

for _, tile in repaired_tiles_df.iterrows():
    params = {
        "parameters": "T2M_MAX",
        "community": "AG",
        "longitude-min": round(float(tile["longitude_min"]), 4),
        "longitude-max": round(float(tile["longitude_max"]), 4),
        "latitude-min": round(float(tile["latitude_min"]), 4),
        "latitude-max": round(float(tile["latitude_max"]), 4),
        "start": "20130101",
        "end": "20130103",
        "format": "JSON",
        "time-standard": "UTC"
    }

    status = None
    error_message = None
    returned_count = 0
    attempts_used = 0

    for attempt in range(1, 4):
        attempts_used = attempt
        try:
            response = requests.get(
                regional_url, params=params, timeout=90
            )
            status = response.status_code

            if not response.ok:
                error_message = response.text[:2000]
                if attempt < 3:
                    time.sleep(2 * attempt)
                    continue
                break

            payload = response.json()
            features = payload.get("features", [])

            if not features:
                error_message = "Successful HTTP response but no features returned"
                if attempt < 3:
                    time.sleep(2 * attempt)
                    continue
                break

            for feature in features:
                longitude, latitude, *elevation = (
                    feature["geometry"]["coordinates"]
                )
                values = feature["properties"]["parameter"]["T2M_MAX"]

                grid_records.append({
                    "tile_id": tile["tile_id"],
                    "grid_latitude": float(latitude),
                    "grid_longitude": float(longitude),
                    "grid_elevation_m": (
                        float(elevation[0]) if elevation else np.nan
                    ),
                    "T2M_MAX_20130101": values.get("20130101"),
                    "T2M_MAX_20130102": values.get("20130102"),
                    "T2M_MAX_20130103": values.get("20130103")
                })

            returned_count = len(features)
            error_message = None
            break

        except Exception as exc:
            error_message = f"{type(exc).__name__}: {exc}"
            if attempt < 3:
                time.sleep(2 * attempt)

    request_log.append({
        "tile_id": tile["tile_id"],
        "http_status": status,
        "attempts_used": attempts_used,
        "grid_points_returned": returned_count,
        "error": error_message
    })

    print(
        f'{tile["tile_id"]}: status={status}, '
        f'points={returned_count}, attempts={attempts_used}'
    )
    time.sleep(0.3)

grid_raw_repaired = pd.DataFrame(grid_records)
request_audit_repaired = pd.DataFrame(request_log)

raw_path = audit_dir / "nasa_power_regional_grid_raw_sample_repaired.csv"
log_path = audit_dir / "nasa_power_regional_grid_request_audit_repaired.csv"

grid_raw_repaired.to_csv(raw_path, index=False)
request_audit_repaired.to_csv(log_path, index=False)

coord_cols = ["grid_latitude", "grid_longitude"]
value_cols = [
    "T2M_MAX_20130101",
    "T2M_MAX_20130102",
    "T2M_MAX_20130103"
]

unique_coords = grid_raw_repaired[coord_cols].drop_duplicates()
duplicate_rows = grid_raw_repaired[
    grid_raw_repaired.duplicated(coord_cols, keep=False)
]

conflicts = []
for coords, group in duplicate_rows.groupby(coord_cols, dropna=False):
    if (group[value_cols].nunique(dropna=False) > 1).any():
        conflicts.append({
            "grid_latitude": coords[0],
            "grid_longitude": coords[1],
            "tiles": ", ".join(sorted(group["tile_id"].unique()))
        })

print("\nREPAIRED GRID INVENTORY SUMMARY")
print("-" * 60)
print("Tiles planned:", len(repaired_tiles_df))
print("Tiles with errors:", int(request_audit_repaired["error"].notna().sum()))
print("Raw grid records:", len(grid_raw_repaired))
print("Unique grid coordinates:", len(unique_coords))
print("Duplicate records from tile overlaps:",
      len(grid_raw_repaired) - len(unique_coords))
print("Coordinates with conflicting sample values:", len(conflicts))
print("Raw inventory saved:", raw_path)
print("Request audit saved:", log_path)

if request_audit_repaired["error"].notna().any():
    print("\nFAILED REQUESTS")
    display(request_audit_repaired[
        request_audit_repaired["error"].notna()
    ])

if conflicts:
    print("\nCONFLICTING OVERLAP VALUES")
    display(pd.DataFrame(conflicts))

display(request_audit_repaired)
```

```python
# STEP 16: Map eligible district coordinates to the nearest
# NASA POWER regional grid point.
#
# This is a mapping and distance audit only.
# It does NOT download additional weather data or modify source files.

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

project_root = Path.cwd().parent

# ---------------------------------------------------------
# 1. Load the district coordinate crosswalk and grid inventory
# ---------------------------------------------------------
crosswalk_path = (
    project_root
    / "metadata"
    / "apy_district_coordinate_crosswalk.csv"
)

grid_path = (
    project_root
    / "reports"
    / "data_audit"
    / "nasa_power_regional_grid_raw_sample_repaired.csv"
)

crosswalk = pd.read_csv(crosswalk_path)
grid = pd.read_csv(grid_path)

print("Crosswalk shape:", crosswalk.shape)
print("Grid inventory shape:", grid.shape)
print("Crosswalk columns:", crosswalk.columns.tolist())
print("Grid columns:", grid.columns.tolist())

# ---------------------------------------------------------
# 2. Keep only non-synthetic district coordinates
# ---------------------------------------------------------
eligible = crosswalk.loc[
    crosswalk["is_synthetic"].astype(str).str.lower().isin(
        ["false", "0"]
    )
].copy()

eligible["latitude"] = pd.to_numeric(
    eligible["latitude"], errors="coerce"
)
eligible["longitude"] = pd.to_numeric(
    eligible["longitude"], errors="coerce"
)

eligible = eligible.dropna(
    subset=["latitude", "longitude"]
).reset_index(drop=True)

# Ensure the grid inventory has valid coordinates
grid["grid_latitude"] = pd.to_numeric(
    grid["grid_latitude"], errors="coerce"
)
grid["grid_longitude"] = pd.to_numeric(
    grid["grid_longitude"], errors="coerce"
)

grid = grid.dropna(
    subset=["grid_latitude", "grid_longitude"]
).drop_duplicates(
    subset=["grid_latitude", "grid_longitude"]
).reset_index(drop=True)

print("\nEligible non-synthetic district coordinates:", len(eligible))
print("Unique regional grid points:", len(grid))

# ---------------------------------------------------------
# 3. Haversine distance in kilometres
# ---------------------------------------------------------
def haversine_km(lat1, lon1, lat2, lon2):
    """Calculate great-circle distance in km.
    Supports scalar district coordinates and array grid coordinates.
    """
    earth_radius_km = 6371.0088

    # Convert each argument independently to support mixed shapes
    lat1 = np.radians(float(lat1))
    lon1 = np.radians(float(lon1))
    lat2 = np.radians(np.asarray(lat2, dtype=float))
    lon2 = np.radians(np.asarray(lon2, dtype=float))

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    # Clip floating-point rounding errors
    a = np.clip(a, 0, 1)

    return 2 * earth_radius_km * np.arcsin(np.sqrt(a))

# ---------------------------------------------------------
# 4. Find the nearest inventoried grid point for each district
# ---------------------------------------------------------
grid_latitudes = grid["grid_latitude"].to_numpy()
grid_longitudes = grid["grid_longitude"].to_numpy()

nearest_records = []

for _, district in eligible.iterrows():
    distances = haversine_km(
        district["latitude"],
        district["longitude"],
        grid_latitudes,
        grid_longitudes
    )

    nearest_index = int(np.argmin(distances))
    nearest_grid = grid.iloc[nearest_index]

    nearest_records.append({
        "apy_geography_id": district["apy_geography_id"],
        "State": district["State"],
        "District": district["District"],
        "district_latitude": district["latitude"],
        "district_longitude": district["longitude"],
        "coordinate_source": district["coordinate_source"],
        "geography_confidence": district["geography_confidence"],
        "nearest_grid_latitude": nearest_grid["grid_latitude"],
        "nearest_grid_longitude": nearest_grid["grid_longitude"],
        "nearest_grid_elevation_m": nearest_grid.get(
            "grid_elevation_m", np.nan
        ),
        "nearest_grid_distance_km": float(
            distances[nearest_index]
        ),
    })

nearest_grid_mapping = pd.DataFrame(nearest_records)

# ---------------------------------------------------------
# 5. Summarize distance and inspect the farthest matches
# ---------------------------------------------------------
distance_summary = (
    nearest_grid_mapping["nearest_grid_distance_km"]
    .describe(percentiles=[0.50, 0.90, 0.95, 0.99])
    .to_frame("distance_km")
)

print("\nNEAREST-GRID DISTANCE SUMMARY")
display(distance_summary.round(3))

print("\nFARTHEST 15 DISTRICT-TO-GRID MATCHES")
display(
    nearest_grid_mapping
    .sort_values("nearest_grid_distance_km", ascending=False)
    .head(15)
    .round(4)
)

# ---------------------------------------------------------
# 6. Save the mapping audit
# ---------------------------------------------------------
output_path = (
    project_root
    / "reports"
    / "data_audit"
    / "district_to_nasa_power_grid_mapping_audit.csv"
)

output_path.parent.mkdir(parents=True, exist_ok=True)
nearest_grid_mapping.to_csv(output_path, index=False)

print("\nMapping rows:", len(nearest_grid_mapping))
print("Unique mapped district IDs:",
      nearest_grid_mapping["apy_geography_id"].nunique())
print("Saved:", output_path)
```

```python
# STEP 17: Validate regional grid values against the point-API pilot

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

project_root = Path.cwd().parent

grid = pd.read_csv(
    project_root / "reports" / "data_audit"
    / "nasa_power_regional_grid_raw_sample_repaired.csv"
)

mapping = pd.read_csv(
    project_root / "reports" / "data_audit"
    / "district_to_nasa_power_grid_mapping_audit.csv"
)

pilot = pd.read_csv(
    project_root / "data" / "processed"
    / "weather_pilot_2013_2024"
    / "nasa_power_weather_pilot_2013_2024.csv"
)

# Standardize the identifier column for this validation.
pilot = pilot.rename(columns={"geography_id": "apy_geography_id"})

pilot["DATE"] = pd.to_datetime(pilot["DATE"])
sample_dates = pd.to_datetime(
    ["2013-01-01", "2013-01-02", "2013-01-03"]
)

pilot_sample = pilot[pilot["DATE"].isin(sample_dates)].copy()

# Keep only pilot locations that have a district-to-grid mapping.
pilot_ids = pilot_sample["apy_geography_id"].unique()
pilot_mapping = mapping[
    mapping["apy_geography_id"].isin(pilot_ids)
].copy()

# Join each pilot location to its selected regional grid point.
comparison = pilot_mapping.merge(
    grid,
    left_on=["nearest_grid_latitude", "nearest_grid_longitude"],
    right_on=["grid_latitude", "grid_longitude"],
    how="left",
    validate="many_to_one"
)

# Convert the regional sample's wide columns into a long table.
regional_cols = {
    "T2M_MAX_20130101": pd.Timestamp("2013-01-01"),
    "T2M_MAX_20130102": pd.Timestamp("2013-01-02"),
    "T2M_MAX_20130103": pd.Timestamp("2013-01-03"),
}

regional_records = []

for _, row in comparison.iterrows():
    for column, date in regional_cols.items():
        regional_records.append({
            "apy_geography_id": row["apy_geography_id"],
            "DATE": date,
            "regional_T2M_MAX": row[column],
            "grid_latitude": row["nearest_grid_latitude"],
            "grid_longitude": row["nearest_grid_longitude"],
            "grid_distance_km": row["nearest_grid_distance_km"],
        })

regional_sample = pd.DataFrame(regional_records)

# Compare regional API values with point-API values.
validation = regional_sample.merge(
    pilot_sample[
        ["apy_geography_id", "DATE", "T2M_MAX"]
    ],
    on=["apy_geography_id", "DATE"],
    how="left",
    validate="one_to_one"
)

validation["difference_C"] = (
    validation["regional_T2M_MAX"] - validation["T2M_MAX"]
)

validation["match_within_0_01C"] = np.isclose(
    validation["regional_T2M_MAX"],
    validation["T2M_MAX"],
    atol=0.01,
    rtol=0,
    equal_nan=False
)

print("Pilot locations:", pilot_mapping["apy_geography_id"].nunique())
print("Comparisons:", len(validation))
print("Missing regional values:",
      validation["regional_T2M_MAX"].isna().sum())
print("Missing point-API values:",
      validation["T2M_MAX"].isna().sum())
print("Matches within 0.01 °C:",
      validation["match_within_0_01C"].sum())
print("Maximum absolute difference (°C):",
      validation["difference_C"].abs().max())

print("\nLOCATION-LEVEL RESULTS")
display(
    validation.groupby("apy_geography_id", as_index=False).agg(
        days_compared=("DATE", "count"),
        matches=("match_within_0_01C", "sum"),
        max_abs_difference_C=(
            "difference_C",
            lambda x: x.abs().max()
        )
    ).round(4)
)

print("\nALL COMPARISONS")
display(validation.round(4))

output_path = (
    project_root / "reports" / "data_audit"
    / "nasa_power_point_vs_regional_validation.csv"
)
validation.to_csv(output_path, index=False)
print("\nSaved:", output_path)
```

```python

import json
import time
import requests
import pandas as pd
from pathlib import Path
from datetime import datetime

# ----------------------------
# Configuration
# ----------------------------
API_URL = "https://power.larc.nasa.gov/api/temporal/daily/regional"

RAW_DIR = project_root / "data" / "raw" / "weather_regional"
LOG_DIR = project_root / "reports" / "data_audit"

RAW_DIR.mkdir(parents=True, exist_ok=True)
LOG_DIR.mkdir(parents=True, exist_ok=True)

CHECKPOINT_PATH = LOG_DIR / "nasa_power_regional_download_log.csv"

# Start with one small test request.
TEST_TILE_IDS = ["R001"]
TEST_YEARS = [2013]
TEST_PARAMETERS = ["T2M_MAX"]

# ----------------------------
# Resume existing log
# ----------------------------
if CHECKPOINT_PATH.exists():
    download_log = pd.read_csv(CHECKPOINT_PATH)
else:
    download_log = pd.DataFrame(columns=[
        "tile_id", "year", "parameter", "status",
        "attempts", "http_status", "response_size_mb",
        "feature_count", "error", "output_file", "timestamp"
    ])

session = requests.Session()
new_log_rows = []

for tile_id in TEST_TILE_IDS:
    tile = repaired_tiles_df.loc[
        repaired_tiles_df["tile_id"] == tile_id
    ].iloc[0]

    for year in TEST_YEARS:
        start_date = f"{year}0101"
        end_date = f"{year}1231"

        for parameter in TEST_PARAMETERS:
            output_dir = RAW_DIR / parameter / str(year)
            output_dir.mkdir(parents=True, exist_ok=True)

            output_path = output_dir / f"{tile_id}_{parameter}_{year}.json"

            # Skip only if the file exists and a previous successful
            # checkpoint entry exists for this exact request.
            already_successful = (
                output_path.exists()
                and (
                    (download_log["tile_id"] == tile_id)
                    & (download_log["year"] == year)
                    & (download_log["parameter"] == parameter)
                    & (download_log["status"] == "SUCCESS")
                ).any()
            )

            if already_successful:
                print(f"Skipping completed request: {tile_id}, {year}, {parameter}")
                continue

            params = {
                "parameters": parameter,
                "community": "AG",
                "longitude-min": float(tile["longitude_min"]),
                "longitude-max": float(tile["longitude_max"]),
                "latitude-min": float(tile["latitude_min"]),
                "latitude-max": float(tile["latitude_max"]),
                "start": start_date,
                "end": end_date,
                "format": "JSON",
                "time-standard": "UTC",
            }

            status = "FAILED"
            error = ""
            http_status = None
            response_size_mb = None
            feature_count = None
            attempts_used = 0

            for attempt in range(1, 4):
                attempts_used = attempt

                try:
                    response = session.get(
                        API_URL,
                        params=params,
                        timeout=180
                    )
                    http_status = response.status_code
                    response_size_mb = round(
                        len(response.content) / (1024 ** 2), 3
                    )

                    if response.status_code == 200:
                        payload = response.json()
                        features = payload.get("features", [])
                        feature_count = len(features)

                        if not features:
                            raise ValueError(
                                "HTTP 200 but response contains no grid features"
                            )

                        # Basic structure check for every feature.
                        for feature in features:
                            values = (
                                feature.get("properties", {})
                                .get("parameter", {})
                                .get(parameter, {})
                            )
                            if not values:
                                raise ValueError(
                                    f"Missing {parameter} values in a grid feature"
                                )

                        output_path.write_text(
                            json.dumps(payload),
                            encoding="utf-8"
                        )

                        status = "SUCCESS"
                        error = ""
                        break

                    error = response.text[:500]

                    # Do not repeatedly retry a clearly invalid request.
                    if response.status_code not in (429, 500, 502, 503, 504):
                        break

                except Exception as exc:
                    error = str(exc)

                if attempt < 3:
                    time.sleep(2 ** attempt)

            row = {
                "tile_id": tile_id,
                "year": year,
                "parameter": parameter,
                "status": status,
                "attempts": attempts_used,
                "http_status": http_status,
                "response_size_mb": response_size_mb,
                "feature_count": feature_count,
                "error": error,
                "output_file": str(output_path) if status == "SUCCESS" else "",
                "timestamp": datetime.now().isoformat(timespec="seconds"),
            }

            new_log_rows.append(row)
            print(
                f"{tile_id} | {year} | {parameter} | "
                f"{status} | HTTP {http_status} | "
                f"{feature_count} features"
            )

            # Save progress immediately after each request.
            if new_log_rows:
                combined_log = pd.concat(
                    [download_log, pd.DataFrame(new_log_rows)],
                    ignore_index=True
                )
                combined_log.to_csv(CHECKPOINT_PATH, index=False)
                download_log = combined_log
                new_log_rows = []

print("\nCheckpoint log:", CHECKPOINT_PATH)
print("Test complete.")
```

```python

import requests
import json
import time
import pandas as pd
from pathlib import Path
from datetime import date, timedelta

API_URL = "https://power.larc.nasa.gov/api/temporal/daily/regional"

RAW_DIR = project_root / "data" / "raw" / "weather_regional_pilot"
RAW_DIR.mkdir(parents=True, exist_ok=True)

pilot_cases = [
    (tile_id, year, parameter)
    for tile_id in ["R001", "R056"]
    for year in [2013, 2020]
    for parameter in ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]
]

# Six? There are 2 tiles × 2 years × 3 variables = 12 requests.
print("Pilot requests:", len(pilot_cases))
print("Tiles: R001, R056 | Years: 2013, 2020")

tile_lookup = repaired_tiles_df.set_index("tile_id")
session = requests.Session()
pilot_results = []

for tile_id, year, parameter in pilot_cases:
    tile = tile_lookup.loc[tile_id]
    start = date(year, 1, 1)
    end = date(year, 12, 31)

    expected_dates = {
        (start + timedelta(days=i)).strftime("%Y%m%d")
        for i in range((end - start).days + 1)
    }

    params = {
        "parameters": parameter,
        "community": "AG",
        "longitude-min": float(tile["longitude_min"]),
        "longitude-max": float(tile["longitude_max"]),
        "latitude-min": float(tile["latitude_min"]),
        "latitude-max": float(tile["latitude_max"]),
        "start": start.strftime("%Y%m%d"),
        "end": end.strftime("%Y%m%d"),
        "format": "JSON",
        "time-standard": "UTC",
    }

    output_path = RAW_DIR / f"{tile_id}_{parameter}_{year}.json"
    status = "FAILED"
    error = ""
    http_status = None
    feature_count = 0
    bad_date_points = None

    for attempt in range(1, 4):
        try:
            response = session.get(API_URL, params=params, timeout=180)
            http_status = response.status_code
            response.raise_for_status()
            payload = response.json()
            features = payload.get("features", [])

            if not features:
                raise ValueError("Response contains no grid features")

            bad_points = []
            for i, feature in enumerate(features):
                values = (
                    feature.get("properties", {})
                    .get("parameter", {})
                    .get(parameter, {})
                )
                if set(values.keys()) != expected_dates:
                    bad_points.append(i)

            if bad_points:
                raise ValueError(
                    f"{len(bad_points)} grid points have incorrect date coverage"
                )

            output_path.write_text(
                json.dumps(payload),
                encoding="utf-8"
            )

            status = "SUCCESS"
            feature_count = len(features)
            bad_date_points = 0
            break

        except Exception as exc:
            error = str(exc)
            if attempt < 3:
                time.sleep(2 ** attempt)

    pilot_results.append({
        "tile_id": tile_id,
        "year": year,
        "parameter": parameter,
        "status": status,
        "http_status": http_status,
        "feature_count": feature_count,
        "expected_days": len(expected_dates),
        "bad_date_points": bad_date_points,
        "error": error,
        "output_file": str(output_path) if status == "SUCCESS" else "",
    })

    print(
        f"{tile_id} | {year} | {parameter} | "
        f"{status} | HTTP {http_status} | features={feature_count}"
    )

    time.sleep(1)

pilot_results_df = pd.DataFrame(pilot_results)

print("\nPilot summary:")
display(pilot_results_df)
print("\nSuccessful:", (pilot_results_df["status"] == "SUCCESS").sum())
print("Failed:", (pilot_results_df["status"] != "SUCCESS").sum())
```

```python

import json
import time
import requests
import pandas as pd
from pathlib import Path
from datetime import date, timedelta
from requests.exceptions import RequestException

# =========================================================
# Configuration
# =========================================================
API_URL = "https://power.larc.nasa.gov/api/temporal/daily/regional"

RAW_DIR = project_root / "data" / "raw" / "weather_regional"
LOG_PATH = (
    project_root / "reports" / "data_audit"
    / "nasa_power_regional_download_log.csv"
)

YEARS = list(range(2013, 2025))
PARAMETERS = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]

MAX_ATTEMPTS = 3
REQUEST_PAUSE_SECONDS = 1.5
TIMEOUT_SECONDS = 180
COORD_TOLERANCE = 0.01  # degrees; allow small boundary rounding differences

RAW_DIR.mkdir(parents=True, exist_ok=True)
LOG_PATH.parent.mkdir(parents=True, exist_ok=True)

tiles = repaired_tiles_df.set_index("tile_id", drop=False)

# Stable, unique acquisition plan.
request_plan = pd.MultiIndex.from_product(
    [repaired_tiles_df["tile_id"].tolist(), YEARS, PARAMETERS],
    names=["tile_id", "year", "parameter"]
).to_frame(index=False)

assert repaired_tiles_df["tile_id"].is_unique
assert len(request_plan) == 2016

# =========================================================
# Load existing checkpoint log and retain latest row per key
# =========================================================
log_columns = [
    "tile_id", "year", "parameter", "status", "attempts",
    "http_status", "response_size_mb", "feature_count",
    "expected_days", "bad_date_points", "error",
    "output_file", "timestamp"
]

if LOG_PATH.exists():
    log_df = pd.read_csv(LOG_PATH)

    for col in log_columns:
        if col not in log_df.columns:
            log_df[col] = None

    log_df = log_df[log_columns].drop_duplicates(
        subset=["tile_id", "year", "parameter"],
        keep="last"
    )
else:
    log_df = pd.DataFrame(columns=log_columns)

def save_log():
    """Persist one latest status per request."""
    global log_df
    log_df = log_df.drop_duplicates(
        subset=["tile_id", "year", "parameter"],
        keep="last"
    )
    log_df.to_csv(LOG_PATH, index=False)

def expected_dates_for_year(year):
    start = date(year, 1, 1)
    end = date(year, 12, 31)
    return {
        (start + timedelta(days=i)).strftime("%Y%m%d")
        for i in range((end - start).days + 1)
    }

def validate_payload(payload, tile, year, parameter):
    """Validate features, coordinates and complete daily date coverage."""
    features = payload.get("features", [])
    if not features:
        raise ValueError("No grid features returned")

    expected_dates = expected_dates_for_year(year)
    bad_date_points = 0

    lat_min = float(tile["latitude_min"])
    lat_max = float(tile["latitude_max"])
    lon_min = float(tile["longitude_min"])
    lon_max = float(tile["longitude_max"])

    for index, feature in enumerate(features):
        coordinates = feature.get("geometry", {}).get("coordinates", [])

        if len(coordinates) < 2:
            raise ValueError(f"Feature {index} has invalid coordinates")

        lon, lat = float(coordinates[0]), float(coordinates[1])

        if not (
            lon_min - COORD_TOLERANCE <= lon <= lon_max + COORD_TOLERANCE
            and lat_min - COORD_TOLERANCE <= lat <= lat_max + COORD_TOLERANCE
        ):
            raise ValueError(
                f"Feature coordinate outside tile bounds: {(lon, lat)}"
            )

        values = (
            feature.get("properties", {})
            .get("parameter", {})
            .get(parameter, {})
        )

        if not values:
            raise ValueError(f"Feature {index} has no {parameter} values")

        if set(values.keys()) != expected_dates:
            bad_date_points += 1

    if bad_date_points:
        raise ValueError(
            f"{bad_date_points} features have incomplete or unexpected dates"
        )

    return len(features), len(expected_dates), bad_date_points

# =========================================================
# Acquisition loop
# =========================================================
session = requests.Session()
total = len(request_plan)
success_count = 0
skipped_count = 0
failed_count = 0

for request_number, request_row in enumerate(
    request_plan.itertuples(index=False), start=1
):
    tile_id = request_row.tile_id
    year = int(request_row.year)
    parameter = request_row.parameter
    tile = tiles.loc[tile_id]

    output_path = (
        RAW_DIR / parameter / str(year)
        / f"{tile_id}_{parameter}_{year}.json"
    )
    output_path.parent.mkdir(parents=True, exist_ok=True)

    # Resume only when a prior success and a file both exist.
    existing = log_df[
        (log_df["tile_id"] == tile_id)
        & (pd.to_numeric(log_df["year"], errors="coerce") == year)
        & (log_df["parameter"] == parameter)
        & (log_df["status"] == "SUCCESS")
    ]

    if not existing.empty and output_path.exists():
        skipped_count += 1
        continue

    start_date = date(year, 1, 1)
    end_date = date(year, 12, 31)

    params = {
        "parameters": parameter,
        "community": "AG",
        "longitude-min": float(tile["longitude_min"]),
        "longitude-max": float(tile["longitude_max"]),
        "latitude-min": float(tile["latitude_min"]),
        "latitude-max": float(tile["latitude_max"]),
        "start": start_date.strftime("%Y%m%d"),
        "end": end_date.strftime("%Y%m%d"),
        "format": "JSON",
        "time-standard": "UTC",
    }

    status = "FAILED"
    error = ""
    http_status = None
    response_size_mb = None
    feature_count = None
    expected_days = len(expected_dates_for_year(year))
    bad_date_points = None
    attempts_used = 0

    for attempt in range(1, MAX_ATTEMPTS + 1):
        attempts_used = attempt

        try:
            response = session.get(
                API_URL,
                params=params,
                timeout=TIMEOUT_SECONDS
            )
            http_status = response.status_code
            response_size_mb = round(
                len(response.content) / (1024 ** 2), 3
            )

            if response.status_code != 200:
                error = response.text[:1000]

                # Retry rate limits and server-side errors only.
                if response.status_code in (429, 500, 502, 503, 504):
                    if attempt < MAX_ATTEMPTS:
                        time.sleep(2 ** attempt)
                        continue
                break

            payload = response.json()
            feature_count, expected_days, bad_date_points = validate_payload(
                payload, tile, year, parameter
            )

            # Save only after full validation passes.
            temp_path = output_path.with_suffix(".json.tmp")
            temp_path.write_text(
                json.dumps(payload),
                encoding="utf-8"
            )
            temp_path.replace(output_path)

            status = "SUCCESS"
            error = ""
            break

        except (RequestException, ValueError, json.JSONDecodeError) as exc:
            error = str(exc)

            if attempt < MAX_ATTEMPTS:
                time.sleep(2 ** attempt)

    new_row = {
        "tile_id": tile_id,
        "year": year,
        "parameter": parameter,
        "status": status,
        "attempts": attempts_used,
        "http_status": http_status,
        "response_size_mb": response_size_mb,
        "feature_count": feature_count,
        "expected_days": expected_days,
        "bad_date_points": bad_date_points,
        "error": error,
        "output_file": str(output_path) if status == "SUCCESS" else "",
        "timestamp": pd.Timestamp.now().isoformat(timespec="seconds"),
    }

    log_df = log_df[
        ~(
            (log_df["tile_id"] == tile_id)
            & (pd.to_numeric(log_df["year"], errors="coerce") == year)
            & (log_df["parameter"] == parameter)
        )
    ]
    log_df = pd.concat(
        [log_df, pd.DataFrame([new_row])],
        ignore_index=True
    )
    save_log()

    if status == "SUCCESS":
        success_count += 1
    else:
        failed_count += 1

    print(
        f"[{request_number}/{total}] "
        f"{tile_id} | {year} | {parameter} | "
        f"{status} | HTTP {http_status} | "
        f"features={feature_count} | attempt={attempts_used}"
    )

    time.sleep(REQUEST_PAUSE_SECONDS)

print("\nACQUISITION RUN FINISHED")
print("New successes:", success_count)
print("Skipped previously successful requests:", skipped_count)
print("Failed requests this run:", failed_count)
print("Checkpoint log:", LOG_PATH)

print("\nLatest status summary:")
display(
    log_df.groupby(["status"], dropna=False)
    .size()
    .rename("request_count")
    .reset_index()
)
```

```python
import json
import pandas as pd
from pathlib import Path
from datetime import date, timedelta

# --------------------------------------------------
# 1. Paths and configuration
# --------------------------------------------------
RAW_DIR = project_root / "data" / "raw" / "weather_regional"
LOG_PATH = (
    project_root / "reports" / "data_audit"
    / "nasa_power_regional_download_log.csv"
)
AUDIT_PATH = (
    project_root / "reports" / "data_audit"
    / "nasa_power_regional_file_audit.csv"
)

YEARS = range(2013, 2025)
PARAMETERS = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]

download_log_df = pd.read_csv(LOG_PATH)

# Derive tile IDs from the actual log; no repaired_tiles_df needed.
TILE_IDS = sorted(download_log_df["tile_id"].dropna().unique())

print("Unique tiles in log:", len(TILE_IDS))
print("Expected requests:", len(TILE_IDS) * len(YEARS) * len(PARAMETERS))

# --------------------------------------------------
# 2. Audit every expected JSON file
# --------------------------------------------------
audit_rows = []

for parameter in PARAMETERS:
    for year in YEARS:
        expected_days = (
            date(year + 1, 1, 1) - date(year, 1, 1)
        ).days

        for tile_id in TILE_IDS:
            file_path = (
                RAW_DIR / parameter / str(year)
                / f"{tile_id}_{parameter}_{year}.json"
            )

            row = {
                "tile_id": tile_id,
                "year": year,
                "parameter": parameter,
                "file_exists": file_path.exists(),
                "valid_json": False,
                "feature_count": 0,
                "unique_coordinates": 0,
                "bad_date_points": None,
                "error": None,
            }

            if not file_path.exists():
                row["error"] = "FILE_MISSING"
                audit_rows.append(row)
                continue

            try:
                with open(file_path, "r", encoding="utf-8") as f:
                    payload = json.load(f)

                features = payload.get("features", [])
                row["valid_json"] = True
                row["feature_count"] = len(features)

                coordinates = set()
                bad_date_points = 0

                expected_dates = {
                    (date(year, 1, 1) + timedelta(days=i)).strftime("%Y%m%d")
                    for i in range(expected_days)
                }

                for feature in features:
                    coords = feature.get("geometry", {}).get("coordinates", [])
                    if len(coords) >= 2:
                        coordinates.add((coords[0], coords[1]))

                    values = (
                        feature.get("properties", {})
                        .get("parameter", {})
                        .get(parameter, {})
                    )

                    actual_dates = set(values.keys())
                    if actual_dates != expected_dates:
                        bad_date_points += 1

                row["unique_coordinates"] = len(coordinates)
                row["bad_date_points"] = bad_date_points

                if not features:
                    row["error"] = "NO_FEATURES"
                elif bad_date_points > 0:
                    row["error"] = "DATE_MISMATCH"
                elif len(coordinates) != len(features):
                    row["error"] = "DUPLICATE_OR_INVALID_COORDINATES"

            except Exception as exc:
                row["error"] = f"JSON_READ_ERROR: {exc}"

            audit_rows.append(row)

# --------------------------------------------------
# 3. Save audit and summarize
# --------------------------------------------------
audit_df = pd.DataFrame(audit_rows)
audit_df.to_csv(AUDIT_PATH, index=False)

print("\n--- FILE-LEVEL AUDIT SUMMARY ---")
print("Expected files:", len(audit_df))
print("Files present:", int(audit_df["file_exists"].sum()))
print("Valid JSON:", int(audit_df["valid_json"].sum()))
print("Files with date issues:", int((audit_df["bad_date_points"].fillna(0) > 0).sum()))
print("Files with errors:", int(audit_df["error"].notna().sum()))
print("Audit saved to:", AUDIT_PATH)

issues_df = audit_df[audit_df["error"].notna()]

if issues_df.empty:
    print("\nNo file-level issues detected.")
else:
    print("\n--- ISSUES FOUND ---")
    display(issues_df.head(30))
```

```python

import json
import numpy as np
import pandas as pd
from pathlib import Path

RAW_DIR = project_root / "data" / "raw" / "weather_regional"
AUDIT_DIR = project_root / "reports" / "data_audit"
AUDIT_DIR.mkdir(parents=True, exist_ok=True)

YEARS = range(2013, 2025)
PARAMETERS = ["PRECTOTCORR", "T2M_MAX", "T2M_MIN"]
TILE_IDS = sorted(download_log_df["tile_id"].dropna().unique())

value_audit_rows = []
coordinate_audit_rows = []

for parameter in PARAMETERS:
    for year in YEARS:
        for tile_id in TILE_IDS:
            file_path = (
                RAW_DIR / parameter / str(year)
                / f"{tile_id}_{parameter}_{year}.json"
            )

            with open(file_path, "r", encoding="utf-8") as f:
                payload = json.load(f)

            for feature in payload.get("features", []):
                coords = feature.get("geometry", {}).get("coordinates", [])
                lon, lat = coords[0], coords[1]

                coordinate_audit_rows.append({
                    "tile_id": tile_id,
                    "year": year,
                    "parameter": parameter,
                    "latitude": lat,
                    "longitude": lon,
                    "elevation_m": coords[2] if len(coords) > 2 else np.nan,
                })

                values = (
                    feature.get("properties", {})
                    .get("parameter", {})
                    .get(parameter, {})
                )

                numeric_values = pd.to_numeric(
                    pd.Series(values, dtype="object"),
                    errors="coerce"
                )

                value_audit_rows.append({
                    "tile_id": tile_id,
                    "year": year,
                    "parameter": parameter,
                    "latitude": lat,
                    "longitude": lon,
                    "date_count": len(values),
                    "missing_values": int(numeric_values.isna().sum()),
                    "sentinel_minus_999": int((numeric_values == -999).sum()),
                    "minimum": numeric_values.min(),
                    "maximum": numeric_values.max(),
                    "negative_values": int((numeric_values < 0).sum()),
                    "non_finite_values": int(
                        (~np.isfinite(numeric_values.dropna())).sum()
                    ),
                })

value_audit_df = pd.DataFrame(value_audit_rows)
coordinate_audit_df = pd.DataFrame(coordinate_audit_rows)

# Save detailed audit reports
value_audit_path = AUDIT_DIR / "nasa_power_weather_value_audit.csv"
coordinate_audit_path = AUDIT_DIR / "nasa_power_coordinate_consistency_audit.csv"

value_audit_df.to_csv(value_audit_path, index=False)
coordinate_audit_df.to_csv(coordinate_audit_path, index=False)

# Check whether the same coordinates recur consistently across years
coordinate_consistency = (
    coordinate_audit_df.groupby(["tile_id", "parameter"])
    .agg(
        unique_coordinates=("latitude", lambda s: len(s)),
        unique_latitudes=("latitude", "nunique"),
        unique_longitudes=("longitude", "nunique"),
    )
    .reset_index()
)

# Parameter-specific screening limits for review, not automatic deletion
limits = {
    "PRECTOTCORR": {"min": 0, "max": 500},
    "T2M_MAX": {"min": -10, "max": 60},
    "T2M_MIN": {"min": -15, "max": 45},
}

value_audit_df["outside_screening_range"] = value_audit_df.apply(
    lambda r: (
        pd.notna(r["minimum"])
        and (
            r["minimum"] < limits[r["parameter"]]["min"]
            or r["maximum"] > limits[r["parameter"]]["max"]
        )
    ),
    axis=1
)

print("Files represented in value audit:", len(value_audit_df))
print("Missing values:", int(value_audit_df["missing_values"].sum()))
print("Sentinel -999 values:", int(value_audit_df["sentinel_minus_999"].sum()))
print("Non-finite values:", int(value_audit_df["non_finite_values"].sum()))
print("Values outside broad screening ranges:",
      int(value_audit_df["outside_screening_range"].sum()))
print("Coordinate groups with multiple latitudes:",
      int((coordinate_consistency["unique_latitudes"] > 1).sum()))
print("Coordinate groups with multiple longitudes:",
      int((coordinate_consistency["unique_longitudes"] > 1).sum()))

print("\nValue audit saved to:", value_audit_path)
print("Coordinate audit saved to:", coordinate_audit_path)

display(
    value_audit_df[value_audit_df["outside_screening_range"]].head(20)
)
```

```python

import json
import requests
import pandas as pd
from pathlib import Path

RAW_DIR = project_root / "data" / "raw" / "weather_regional"
AUDIT_DIR = project_root / "reports" / "data_audit"

# Find the 10 largest daily precipitation values across the regional files.
extremes = []

for year in range(2013, 2025):
    for tile_id in TILE_IDS:
        path = RAW_DIR / "PRECTOTCORR" / str(year) / f"{tile_id}_PRECTOTCORR_{year}.json"

        with open(path, "r", encoding="utf-8") as f:
            payload = json.load(f)

        for feature in payload.get("features", []):
            coords = feature["geometry"]["coordinates"]
            values = feature["properties"]["parameter"]["PRECTOTCORR"]

            for day, value in values.items():
                if value is not None and value != -999:
                    extremes.append({
                        "tile_id": tile_id,
                        "year": year,
                        "date": day,
                        "latitude": coords[1],
                        "longitude": coords[0],
                        "regional_precip_mm": float(value),
                    })

extremes_df = (
    pd.DataFrame(extremes)
    .sort_values("regional_precip_mm", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

print("Ten largest regional daily precipitation values:")
display(extremes_df)

# Compare the largest five against the point API.
comparison_rows = []

for _, row in extremes_df.head(5).iterrows():
    day = str(row["date"])
    lat = float(row["latitude"])
    lon = float(row["longitude"])

    params = {
        "parameters": "PRECTOTCORR",
        "community": "AG",
        "latitude": lat,
        "longitude": lon,
        "start": day,
        "end": day,
        "format": "JSON",
        "time-standard": "UTC",
    }

    try:
        response = requests.get(
            "https://power.larc.nasa.gov/api/temporal/daily/point",
            params=params,
            timeout=60,
        )
        response.raise_for_status()
        payload = response.json()

        point_values = payload["properties"]["parameter"]["PRECTOTCORR"]
        point_value = point_values.get(day)

        comparison_rows.append({
            "tile_id": row["tile_id"],
            "date": day,
            "latitude": lat,
            "longitude": lon,
            "regional_precip_mm": row["regional_precip_mm"],
            "point_precip_mm": point_value,
            "absolute_difference_mm": (
                abs(float(row["regional_precip_mm"]) - float(point_value))
                if point_value is not None else None
            ),
            "status": "OK" if point_value is not None else "POINT_VALUE_MISSING",
        })

    except Exception as exc:
        comparison_rows.append({
            "tile_id": row["tile_id"],
            "date": day,
            "latitude": lat,
            "longitude": lon,
            "regional_precip_mm": row["regional_precip_mm"],
            "point_precip_mm": None,
            "absolute_difference_mm": None,
            "status": f"REQUEST_ERROR: {exc}",
        })

comparison_df = pd.DataFrame(comparison_rows)

comparison_path = AUDIT_DIR / "nasa_power_extreme_precipitation_validation.csv"
comparison_df.to_csv(comparison_path, index=False)

print("\nRegional API vs point API:")
display(comparison_df)
print("Comparison saved to:", comparison_path)
```
