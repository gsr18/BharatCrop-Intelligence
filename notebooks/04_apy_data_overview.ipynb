# APY Dataset — Data Overview

## Objective

This notebook examines the historical Area, Production and Yield (APY) dataset used in BharatCrop Intelligence. Before cleaning or transforming the data, we need to understand its structure, coverage, variables and limitations.

The analysis will cover:
- Dataset dimensions and column structure
- Data types and sample records
- Agricultural years, states, districts, crops and seasons represented
- Units and interpretation of important agricultural variables
- Initial data-quality observations that need further investigation

The original dataset will remain unchanged. Missing values, duplicate records, inconsistent labels and suspicious values will be handled in the data-cleaning notebook, not here.

```python
from pathlib import Path
import pandas as pd

# Locate the project directory from the current notebook location.
project_root = Path.cwd().parent

# Path to the merged APY dataset created during data acquisition.
apy_path = (
    project_root
    / "data"
    / "processed"
    / "historical_food_grains_apy_2013-14_to_2024-25.csv"
)

# Check that the expected file exists before trying to read it.
if not apy_path.exists():
    raise FileNotFoundError(f"APY dataset not found: {apy_path}")

# Load the dataset for inspection. The source file is not modified.
apy_df = pd.read_csv(apy_path)

print("Dataset loaded successfully.")
print(f"File: {apy_path}")
print(f"Rows: {apy_df.shape[0]:,}")
print(f"Columns: {apy_df.shape[1]}")
```

```python
# Check how the APY data is organized and which columns need attention.
print("Column names:")
print(apy_df.columns.tolist())

print("\nDataset information:")
apy_df.info()
```

```python
# View a few records to get a feel for the data.
display(apy_df.head(10))
```

```python
# Identify the agricultural years covered by the dataset.
year_columns = [
    col for col in apy_df.columns
    if "year" in col.lower()
]

print("Potential year columns:", year_columns)

for col in year_columns:
    print(f"\nValues in {col}:")
    print(apy_df[col].dropna().astype(str).drop_duplicates().sort_values().tolist()[:30])
```

```python
# The agricultural year is part of each measurement column's name.
year_columns = [
    col for col in apy_df.columns
    if any(metric in col.lower() for metric in ["area-", "production-", "yield-"])
]

print("Number of agricultural measurement columns:", len(year_columns))
print("\nMeasurement columns:")
print(year_columns)
```

```python
# These columns describe where and what each agricultural record represents.
category_columns = ["State", "District", "Crop", "Season"]

print("Unique values in each category:\n")

for col in category_columns:
    print(f"{col}: {apy_df[col].nunique(dropna=True)} unique values")
    print(apy_df[col].dropna().unique()[:15].tolist())
    print()
```

```python
# Extract the years represented in the measurement columns.
apy_years = sorted({
    col.split("-")[1] + "-" + col.split("-")[2]
    for col in year_columns
})

print("Agricultural years covered:")
print(apy_years)

print(f"\nFirst year: {apy_years[0]}")
print(f"Latest year: {apy_years[-1]}")
print(f"Total years: {len(apy_years)}")
```

```python
# Check whether the records have missing values in their identifying fields.
print("Missing values in category columns:")
display(apy_df[category_columns].isna().sum().to_frame("Missing values"))

# Check whether any rows are repeated across all identifying fields.
print("\nDuplicate records based on location, crop and season:")
print(apy_df.duplicated(subset=category_columns).sum())
```

```python
# Check the data types and missing values across all columns.
overview = pd.DataFrame({
    "Data Type": apy_df.dtypes.astype(str),
    "Missing Values": apy_df.isna().sum(),
    "Missing (%)": (apy_df.isna().mean() * 100).round(2)
})

display(overview)
```

```python
# Get a quick summary of the area, production and yield measurements.
display(apy_df[year_columns].describe().T.round(2))
```

```python
# Summarize missing values by year and measurement type.
missing_rows = []

for year in apy_years:
    for metric in ["Area", "Production", "Yield"]:
        col = f"{metric}-{year}"

        missing_rows.append({
            "Year": year,
            "Metric": metric,
            "Missing Count": apy_df[col].isna().sum(),
            "Missing (%)": round(apy_df[col].isna().mean() * 100, 2)
        })

missing_summary = pd.DataFrame(missing_rows)

display(
    missing_summary.pivot(
        index="Year",
        columns="Metric",
        values="Missing (%)"
    )
)
```

```python
# Zero values can mean different things from missing values,
# so measure their frequency separately.
zero_rows = []

for col in year_columns:
    zero_rows.append({
        "Column": col,
        "Zero Count": (apy_df[col] == 0).sum(),
        "Zero (%)": round((apy_df[col] == 0).mean() * 100, 2)
    })

zero_summary = pd.DataFrame(zero_rows)

display(zero_summary)
```

```python
# Check how zero values in Area, Production, and Yield relate
# to one another in the latest agricultural year.

latest_year = apy_years[-1]

area_col = f"Area-{latest_year}"
production_col = f"Production-{latest_year}"
yield_col = f"Yield-{latest_year}"

zero_patterns = pd.Series(
    "Other combinations",
    index=apy_df.index,
    dtype="object"
)

zero_patterns.loc[
    apy_df[area_col].eq(0) & apy_df[production_col].eq(0)
] = "Area = 0 and Production = 0"

zero_patterns.loc[
    apy_df[area_col].eq(0) & apy_df[production_col].gt(0)
] = "Area = 0, Production > 0"

zero_patterns.loc[
    apy_df[area_col].gt(0) & apy_df[production_col].eq(0)
] = "Area > 0, Production = 0"

zero_patterns.loc[
    apy_df[area_col].gt(0) & apy_df[production_col].gt(0)
] = "Area > 0 and Production > 0"

print(f"Zero-value patterns for {latest_year}:")
display(
    zero_patterns.value_counts()
    .rename_axis("Pattern")
    .to_frame("Record Count")
)

print("\nYield = 0 while Area and Production are both positive:")
display(
    apy_df.loc[
        apy_df[area_col].gt(0)
        & apy_df[production_col].gt(0)
        & apy_df[yield_col].eq(0),
        ["State", "District", "Crop", "Season",
         area_col, production_col, yield_col]
    ].head(10)
)
```

```python
# Inspect records where Area and Production have inconsistent zero values.
# These are candidates for investigation, not automatic errors.

inconsistent_records = apy_df.loc[
    (
        apy_df[area_col].eq(0)
        & apy_df[production_col].gt(0)
    )
    |
    (
        apy_df[area_col].gt(0)
        & apy_df[production_col].eq(0)
    ),
    [
        "State", "District", "Crop", "Season",
        area_col, production_col, yield_col
    ]
]

print(f"Total records requiring investigation: {len(inconsistent_records)}")

display(inconsistent_records.head(20))
```

```python
# Count inconsistent Area and Production combinations for every year.

consistency_rows = []

for year in apy_years:
    area = apy_df[f"Area-{year}"]
    production = apy_df[f"Production-{year}"]

    consistency_rows.append({
        "Year": year,
        "Area = 0, Production > 0": (
            area.eq(0) & production.gt(0)
        ).sum(),
        "Area > 0, Production = 0": (
            area.gt(0) & production.eq(0)
        ).sum()
    })

consistency_summary = pd.DataFrame(consistency_rows)
display(consistency_summary)
```

## Key Findings and Data Limitations

- The dataset contains 31,356 records and 40 columns, covering 34 states and territories, 748 district labels, 19 crop categories, four seasons, and 12 agricultural years from 2013–14 to 2024–25.
- The dataset is in wide format, with separate Area, Production, and Yield columns for each agricultural year.
- The identifier columns contain no missing values, and no duplicate combinations were found using State, District, Crop, and Season.
- Missing values are substantial in the annual agricultural measures. Missingness generally decreases from approximately 33–34% in 2013–14 to approximately 19–20% in 2024–25.
- Zero values are common in Area and Production, but rare in Yield. Missing values and zero values must be treated separately.
- Area–Production zero inconsistencies occur across all 12 years and require investigation during cleaning.
- Some Yield values are unusually large and require validation before they are used in analysis or machine learning.
- Crop categories include both individual crops and aggregate categories, such as Total Food Grains and Total Pulses. These must be distinguished to avoid double-counting during analysis.
- The source measurement units, reporting conventions, and interpretation of unusual values must be verified before applying cleaning rules.

### Conclusion

The dataset provides broad historical coverage for agricultural analysis, but its missingness, zero-value patterns, extreme values, and aggregate crop categories require careful validation. This overview does not modify the original dataset. All cleaning decisions will be documented separately in the next notebook.
